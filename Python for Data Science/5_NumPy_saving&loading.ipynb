**Stacking**
- Stacking refers to joining or combining multiple arrays together along a new or an existing axis to form a single, larger array.
  
| method | description | syntax / example |
| :--- | :--- | :--- |
| np.hstack() | Combines arrays horizontally (column-wise) by adding new columns.<br>• Requires matching number of rows. | np.hstack(arr1,arr2) |
| np.vstack() | Combines arrays vertically (row-wise) by adding new rows.<br>• Requires matching number of columns. | np.vstack((arr1, arr2)) |
| np.row_stack() | Same as `np.vstack()`. | np.row_stack((arr1, arr2)) |
| np.column_stack() | Stacks 1D arrays side-by-side as columns into a 2D matrix (automatically turning 1D vectors into column vectors). | np.column_stack((arr1, arr2)) |

```python
# Stacking
#Vertical(Row) stack; should have the same number of columns
array3=np.array([[1,1,1,1],[2,2,2,2]])
print(f'array3 is \n{array3}\nand has a shape of {array3.shape}')
array4=np.array([3,3,3,3])
print(f'array4 is \n{array4}\nand has a shape of {array4.shape}')
array5=np.vstack([array3,array4])
print(array5,'\n')
```

```python

#Horizontal(Column) stack: should have same number of rows 
print(f'array3 is \n{array3}\nand has a shape of {array3.shape}')
array6=np.array([[3],[4]])
print(f'array6 is \n{array6}\nand has a shape of {array6.shape}')
array7=np.hstack([array3,array6])
print(array7,'\n')
```

```python
# Column stack: stacks 1d arrays into 2d
array8=np.array([1,2,3,4,5])
print(f'array8 is\n{array8}\n')
array9=np.array([6,7,8,9,10])
print(f'array9 is\n{array9}\n')
array10=np.column_stack((array8,array9))
print(f'np.column_stack((array8,array9)) stacks array8 and array9 into a 2d matrix\n{array10}\n')
```

**Saving NumPy Files**

* Opening a saved 1D array with a text editor shows a column of values.
* Two 1D arrays saved in a single file appear as 2 long lines when opened with a text editor.
* This can be solved by stacking.
* The `header` parameter enables us to add headers to text data we are saving.
* NumPy text files can be managed using standard Python file handling (`open()`, `close()`, or `with open()`).
* The `file.flush()` command enables continuous writing to files despite errors.

| File format | Description |
| :--- | :--- |
| .npy | Standard binary file format in NumPy for persisting a single arbitrary NumPy array. |
| .npz | A Zip archive containing multiple `.npy` files (combines multiple arrays into one file).<br>• Can be compressed or uncompressed. |

| Method | Description |
| :--- | :--- |
| np.save() | Save an array to a binary file in NumPy `.npy` format. |
| np.savez() | Save several arrays into a single file in uncompressed `.npz` format. |
| np.savez_compressed() | Save several arrays into a single file in compressed `.npz` format. |
| np.load() | Load arrays or pickled objects from `.npy`, `.npz`, or pickled files. |
| np.savetxt() | Save an array into `.txt` format. |
| np.loadtxt() | Open text files into arrays. |

```python
# Create a sample array and save it
arr11=np.arange(0,200,10)
np.save('arr11_saved', arr=arr11)
print(arr11)
```

```python
# Loading numpy files
np.load('arr11_saved.npy')
```
