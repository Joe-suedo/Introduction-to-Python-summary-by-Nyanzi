# MEASURES OF LOCATION OF DATA(Percentile, Quartiles and Mean)
* Percentiles, Quartiles and the Median are measures of the location of data.
*  THey indicate the relative standing position of a data value within a dataset.

### Shared Properties

| Property | Description |
| :--- | :--- |
| Outlier Resistance | Highly resistant to extreme values compared to the mean, as they are determined by positional ranking rather than numerical magnitude. |
| Directional Scale | Positional values directly reflect data magnitude: lower positions correspond to smaller data values, whereas higher positions correspond to larger values. |
| Sample Presence (Interpolation) | Positional values  may or may not correspond to an actual observed data value in the sample, often falling between points or requiring averaging. |
| Context-Dependent Evaluation & Reporting Standard | Whether a specific value is considered "good" or "bad" depends entirely on the domain context:<br>• Lower is desirable: Metrics like reaction times, completion times, and error rates.<br>• Higher is desirable: Metrics like income levels, test scores, and performance efficiency.<br>•Neutral: Many descriptive scenarios carry no value judgment at all.<br><br> *Reporting Standard:* A complete interpretation requires:<br>1. The specific domain context.<br>2. The exact numerical data value.<br>3. The proportion of observations falling strictly above and below that threshold. |

### 1. Percentiles ($P_k$)
* Statistical values that partition an ordered dataset into 100 equal parts.
* The $k^{\text{th}}$ percentile represents the numerical value below which $k\%$ of the observations lie.

### 2. Quartiles ($Q_1, Q_2, Q_3$)
* Specific percentiles that divide an ordered dataset into four equal segments (quarters).

| Quartile | Percentile Equivalent | Description |
| :--- | :--- | :--- |
| First Quartile ($Q_1$) | 25th percentile | Separates the lowest 25% of data from the upper 75%. |
| Second Quartile ($Q_2$) | 50th percentile | The *Median*; splits the dataset into two equal halves. |
| Third Quartile ($Q_3$) | 75th percentile | Separates the lowest 75% of data from the upper 25%. |

* **Interquartile Range ($\text{IQR}$):**
  * Measures the spread of the middle 50% of the dataset: $\text{IQR} = Q_3 - Q_1$.
  * *Outlier Detection ($1.5 \times \text{IQR}$ Rule):* Identifies potential extreme values using upper and lower boundaries:
    * Lower Outlier Boundary: $< Q_1 - 1.5 \times \text{IQR}$
    * Upper Outlier Boundary: $> Q_3 + 1.5 \times \text{IQR}$

### 3. Median
* The single value separating ordered data into two equal halves (50% above, 50% below).
* Serves as the primary measure of central tendency for skewed distributions or datasets with severe outliers.

---

### Universal Calculation Procedures

* *Prerequisite for All Measures:* Raw data *must* be sorted in ascending order (smallest to largest) prior to any positional calculation.

| Measure | Formula / Calculation Method |
| :--- | :--- |
| Median | • *Odd $n$*: The exact middle observation at position $\frac{n + 1}{2}$.<br>• *Even $n$:* The arithmetic mean of the two central observations $\frac{n}{2}$ and $\frac{n}{2} + 1 $ |
| Quartiles ($Q_1, Q_2, Q_3$) | 1. Calculate the median ($Q_2$) to split the dataset into lower and upper halves.<br>2. $Q_1$ = Median of the lower half of data.<br>3. $Q_3$ = Median of the upper half of data. |
| $k^{\text{th}}$ Percentile Value | 1. Calculate position index: $i = \frac{(n + 1)k}{100}$ (or standard variant $i = \frac{n \cdot k}{100}$).<br>2. If $i$ is an integer: Select the value at the $i^{\text{th}}$ position.<br>3. If $i$ is non-integer: Average the values at the nearest integer positions $\lfloor i \rfloor$ and $\lceil i \rceil$ (or linearly interpolate). |
| Percentile Rank of a Data Value | Formula: $\text{Percentile Rank} = \left(\frac{x + 0.5y}{n}\right) \times 100$<br>• $x$: Count of observations strictly less than the target value.<br>• $y$: Count of observations equal to the target value.<br>• $n$: Total sample size. |

### Box Plots (Box-and-Whisker Plots)

* Provide a clear visual representation of data concentration, spread, and symmetry.
* Display how far extreme values (outliers) lie from the bulk of the data.
* Constructed using five fundamental positional values:
  1. Minimum data value
  2. First Quartile ($Q_1$)
  3. Median ($Q_2$)
  4. Third Quartile ($Q_3$)
  5. Maximum data value

**Construction Rules**

| Part | Description |
| :--- | :--- |
| Axis & Layout | Drawn along a scaled horizontal or vertical number line alongside a rectangular box.|
| The Box| Approximately the middle 50% of the data falls inside the box.<br>Ends are defined by the first quartile ($Q_1$) and the third quartile ($Q_3$).<br>The Length  represents the Interquartile Range ($\text{IQR}$) |
| The Median | Indicated inside the box as a line or marker, positioned between $Q_1$ and $Q_3$ (unless it coincides with either quartile).|
| The Whiskers | Extend outward from the ends of the box to the minimum and maximum data values on either side. |


**Modified Box Plots:**
* Used  When outliers are present
* Whiskers extend only to the last non-outlier data points (or specific fence boundaries like $1.5 \times \text{IQR}$)
* Outliers are plotted individually as points or dots beyond the whiskers.

## MEASURES OF THE CENTER OF DATA (Mean, Median, Mode)

### The Mean (Arithmetic Mean / Average)
* Standard measure of the center of data, typically paired with the median.
  
**Core Property:** 
* The only measure of central tendency that includes every individual value in its calculation.
* The sum of the deviations of each value from the mean is always zero.

**Comparison of the Mean and the Median**

| Mean | Median |
| :--- | :--- |
| Affected by the numerical weight and exact magnitude of every value. | Determined by the ascending hierarchical position/order of data values. |
| The most common measure of the center of data. | Preferred measure of the center in datasets with outliers or skewed distributions. |
| Highly sensitive and easily skewed by extreme outliers. | Resistant and robust against extreme outliers. |

**Calculation of the Mean**

| Data Type | Description | Formula |
| :--- | :--- | :--- |
| Dataset with Unique Values | Sum of all individual values divided by the total count of values. | $\frac{\sum x}{n}$<br>• $x$: Individual data value, $n$: Total number of values. |
| Dataset with Repeated Values | Sum of the products of each value and its frequency divided by the total count of all values, | $\frac {\sum x \cdot f(x)}{n}$<br>• $x$: Data value, $f(x)$: Frequency of that value, $n$: Total number of values. |
| Grouped Frequency Tables | • Grouped data utilize intervals and interval frequencies rather than raw individual values.<br>• An exact mean cannot be calculated; instead, an approximation is computed using class midpoints ($m$). | • Midpoint ($m$) = $\frac{\text{Lower Boundary} + \text{Upper Boundary}}{2}$<br>• Mean = $\frac{\sum m \cdot f}{\sum f}$<br>• $f$: Frequency of the interval. |

**Representation of the Mean**

| Type of Mean | Statistical Symbol & Notation |
| :--- | :--- |
| Sample Mean (Statistic) | $\bar{x}$ (pronounced "x bar") |
| Population Mean (Parameter) | $\mu$ (pronounced "mew") |

*The Law of Large Numbers:* The sample mean ($\bar{x}$) converges toward and closely approximates the true population mean ($\mu$) as the sample size increases within a large population.

---

### Mode
* The most frequently occurring value in a dataset.
* A dataset can contain more than one mode (multimodal) as long as those distinct values share the exact same highest frequency. Specifically, a dataset with two modes is termed *bimodal*.
* Can be calculated for both *qualitative (categorical)* and *quantitative* data.Unlike means and medians which typically require numerical(quantitative) data.

**Symmetrical Distribution**
*  A distribution where a vertical line can divide the graph or figure such that the shape to the left and right of the line are exact mirror images of each other.
* In a perfectly symmetrical (and unimodal) distribution:
  * The *mean*, *median*, and *mode* are equal and coincide at the exact center ($\text{Mean} = \text{Median} = \text{Mode}$).
  * The distribution exhibits zero skewness.

# MEASURES OF SPREAD OF DATA

* Measure the variation in data.
* The most common measure of variation, or spread, is the standard deviation.

### The Standard Deviation

* Provides a numerical measure of the overall amount of variation in a data set.
* It measures how far data values are from their mean.
    * Small when all the data are concentrated close to the mean, exhibiting little variation or spread. 
    * Larger when the data values are more spread out from the mean, exhibiting more variation.
* It is always positive or zero.


**Calculating the Standard Deviation**

| Step | Description | Sample ($s$) | Population ($\sigma$) |
| :--- | :--- | :--- | :--- |
| Step 1: Deviation | Find the difference between each data value and its mean. | $x - \bar{x}$ | $x - \mu$ |
| Step 2: Variance| Compute the average of the squared deviations (divide by $n-1$ for a sample, and $N$ for a population). | $s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$ | $\sigma^2 = \frac{\sum (x - \mu)^2}{N}$ |
| Step 3: Standard Deviation | Take the square root of the variance to return to the original units of data. | $s = \sqrt{\frac{\sum (x - \bar{x})^2}{n - 1}}$ | $\sigma = \sqrt{\frac{\sum (x - \mu)^2}{N}}$ |

**The Z-Score**

* A standardized score used to compare data values across different datasets or distributions.
* Indicates exactly how many standard deviations a given data value is from the mean.
* Computed as the ratio of the difference between a value and the mean to the standard deviation.

| Population Z-Score | Sample Z-Score |
| :--- | :--- |
| Uses population parameters ($\mu$ and $\sigma$) | Uses sample statistics ($\bar{x}$ and $s$) |
| $z = \frac{x - \mu}{\sigma}$ | $z = \frac{x - \bar{x}}{s}$ |

*Empirical / Z-Score Rule (Standard Deviation Method):*
* Data points with a Z-score beyond $\pm 2$ ($|z| > 2$) are considered unusually far from the mean.
* Data points with a Z-score beyond $\pm 3$ ($|z| > 3$) are classified as definitive outliers.
* Best applied to symmetrical, bell-shaped (normally distributed) datasets.

*Comparison of Outlier Methods*

| Metric / Rule | Threshold | Primary Distribution Type | Sensitivity of the Method |
| :--- | :--- | :--- | :--- |
| Z-Score Rule | $\|z\| > 2$ (Unusual)<br>$\|z\| > 3$ (Outlier) | Symmetric / Normal | **Sensitive:** Mean and standard deviation are heavily pulled by extreme values, which distorts the calculation. |
| **$1.5 \times \text{IQR}$ Rule** | $< Q_1 - 1.5(\text{IQR})$<br>or $> Q_3 + 1.5(\text{IQR})$ | Skewed / Non-Normal | **Resistant:** Quartiles and medians depend on positional ranking, so existing outliers cannot distort the threshold boundaries. |

### Types of Variability

| Type | Description |
| :--- | :--- |
| Observational| Arises from differences in the instruments used to measure values or in the people using those instruments. |
| Natural | Arises from the differences that naturally occur because members of a population differ from each other. |
| Induced | An artificially created element of variation that, by definition, was not present naturally. |
| Sample | Occurs when multiple random samples are taken from the same population. |

---

### Sampling Variability of a Statistic

* Measures how much a sample statistic varies from one random sample to another.
* The sampling variability of a statistic is typically quantified by its **standard error**.
* The standard error of the mean is denoted as $\sigma_{\bar{x}} = \frac{\sigma}{\sqrt{n}}$.
