## Recap on Matrices, Vectors

* *Array:* Data structure in computer science and data science to store and manipulate multi-dimensional numerical data efficiently.
* *Vector:* One-dimensional array representing a sequence of numbers (matrix with a single column or row).
* *Matrix:* Two-dimensional data structure consisting of elements (real or complex numbers) arranged in rows and columns.
  
*Order (Dimensions) of a Matrix*
* The dimensions of a matrix are defined as *$m \times n$*, where:
  * *m* represents the number of horizontal rows.
  * *n* represents the number of vertical columns.
 
*Basic Matrix Operations*

* Addition and Subtraction
    * Operations are performed *element-wise* (corresponding elements in identical positions are added or subtracted).
    * Only matrices of the exact same order ($m \times n$) can be added or subtracted.

* Scalar Multiplication
    * Scalar: A single numerical value or magnitude.
    * Every element in the matrix is scaled by the scalar value.
    * The original order and shape of the matrix are fully maintained.

* Matrix Multiplication (Dot Product)
    - Dot product of rows from the first matrix and columns from the second matrix.
    - Number of columns in the first matrix must match the number of rows in the second matrix.
    - Resulting matrix order changes based on input dimensions, yielding an $(m \times p)$ output matrix from $(m \times n)$ and $(n \times p)$ inputs.
    - *Non-commutative:* Matrix multiplication order matters ($A \cdot B \neq B \cdot A$).
    - $c_{ij} = \sum_{k=1}^n a_{ik}b_{kj}$
* Determinants
    * Scalar value computed from the elements of a square matrix that encodes certain properties of the linear transformation described by the matrix.

### Multi dimensional Array Operations

* Operate element-wise on corresponding positions of arrays sharing *identical shapes* by default.(like matrix addition/subtraction)
*Broadcasting* for arrays of different shapes, the smaller array is extended to become compatible to the larger array without copying data.
* Shapes of two arrays are compared element-wise from right to left (starting from the trailing dimensions), meaning the axes are strictly positional and not interchangeable.
*Two dimensions of arrawys are compatible if:
  * they are equal in size(same number of rows or same number of columns)
  * one of them is 1 (single row or column )
* Therefore input arrays can have different numbers of dimensions, and the resulting broadcast array will match the number of dimensions of the array with the largest rank.
* * Dimensions can be explicitly added to an array using `np.newaxis` e.g B[np.newaxis,:]


| operation | description & behavior | syntax |
| :--- | :--- | :--- |
| add | adds corresponding elements of two arrays element-wise (equivalent to `+`). | p.add(arr1, arr2) or arr1+arr2 |
| subtract | subtracts elements element-wise (equivalent to `-`). | np.subtract(arr1, arr2) or arr1-arr2 |
| multiply | multiplies elements element-wise (equivalent to `*`). | np.multiply(arr1, arr2) or arr1*arr2 |
| divide | divides elements element-wise (equivalent to `/`). | np.divide( arr1, arr2) or arr1/arr2 |
| mod | computes the element-wise remainder of division (equivalent to `%`). | np.mod(or arr1,arr2) or arr1%arr2 |

| dot | computes the matrix dot product or inner product of two arrays, which is distinct from element-wise multiplication | arr1.dot(arr2) |
| @ | operator shorthand for matrix multiplication between two arrays | arr1 @ arr2 or np.sum(arr1*arr2) |
| maximum | computes the element-wise maximum of two arrays, returning the larger value at each position | np.maximum(arr1, arr2) |
| minimum | computes the element-wise minimum of two arrays, returning the smaller value at each position | np.minimum(arr1, arr2) |
