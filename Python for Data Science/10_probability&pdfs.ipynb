# BASIC PROBABILITY

## Terminology 

| Term | Description |
| :--- | :--- |
| Probability | A measure of the likelihood or certainty of outcomes resulting from a particular activity. |
| Experiment | An activity or procedure carried out under controlled conditions to observe and measure the certainty of its results. |
| Outcome | A single possible result or observation of an experiment. |
| Sample Space | The complete set or collection of all possible outcomes of an experiment. |
| Event | • Any combination or specific collection of outcomes.<br>• A subset of the sample space.<br>• Represented by uppercase letters (e.g., event $A$).<br>• The probability of event $A$, denoted as $P(A)$, ranges between $0$ and $1$.<br>• Two events share the same probability if they are *equally likely*.<br>• Calculated as: $P(A) = \frac{\text{Number of outcomes for event } A}{\text{Total number of outcomes in sample space}}$.<br>• *Law of Large Numbers:* As the number of experiment repetitions increases, the experimental relative frequency converges closer and closer to the theoretical probability. |
| "Or" Event | An outcome that belongs to event $A$, event $B$, or to both events simultaneously (Union: $A \cup B$). |
| "And" Event | An outcome that belongs to both event $A$ and event $B$ at the same time (Intersection: $A \cap B$). |
| Complement of $A$ | • All possible outcomes in the sample space that are *not* part of event $A$.<br>• Denoted as $A'$ (read as "A prime") or $A^c$. |
| Conditional Probability | • The probability that event $A$ occurs given that event $B$ has already occurred.<br>• Denoted as $P(A \mid B)$ (read as "the probability of $A$ given $B$").<br>• Conditioning on $B$ effectively reduces the sample space to only the outcomes contained within $B$.<br>• **Formula:** $P(A \mid B) = \frac{P(A \text{ and } B)}{P(B)}$, where $P(B) > 0$. |

**Independent and Mutually Exclusive Events**

| Feature | Independent Events | Mutually Exclusive Events |
| :--- | :--- | :--- |
| Definition | The occurrence or knowledge of one event does not affect or change the probability of the other event occurring. | Two or more events that cannot occur at the same time because they share no common outcomes. |
| Mathematical Condition | • $P(A \mid B) = P(A)$<br>• $P(B \mid A) = P(B)$| • $P(A \text{ and } B) = 0$<br>• $P(A \mid B) = 0$ (provided $P(B) > 0$) |

**Multiplication and Addition Rules**
*(Assuming two events $A$ and $B$ defined on the same sample space)*

| Category | Multiplication Rule | Addition Rule |
| :--- | :--- | :--- |
| Formula / Derivation Steps | • Derivation: Starts from the conditional probability definition:<br>  $P(A \mid B) = \frac{P(A \text{ and } B)}{P(B)}$ <br> • *Derived Formula: $P(A \text{ and } B) = P(B) \cdot P(A \mid B)$<br>| $P(A \text{ or } B) = P(A) + P(B) - P(A \text{ and } B)$  |
| Special Application | For Independent Events:<br>Since $P(A \mid B) = P(A)$, the formula simplifies to:<br>$P(A \text{ and } B) = P(A) \cdot P(B)$ | For Mutually Exclusive Events:<br>Since $P(A \text{ and } B) = 0$, the formula simplifies to:<br>$P(A \text{ or } B) = P(A) + P(B)$ |

# Probability Distribution Functions of Random Variables

| Variable Classification | Description |
| :--- | :--- |
| Regular Variable | A fixed, deterministic value or an unknown value solved through standard algebraic equations. |
| Random Variable ($X$) | A variable whose numerical value is determined by the outcome of a random event or experiment.<br>• Can take multiple different values, each governed by an associated probability. |

* *Probability Distribution Function (PDF / PMF):* A mathematical function or rule that describes how total probabilities are distributed across all possible outcomes of a random variable.
* *Cumulative Distribution Function (CDF):* Denoted as $F(x) = P(X \le x)$, it gives the probability that the random variable $X$ takes a value less than or equal to $x$.

**Types of Random Variables**

| Property / Feature | Discrete Random Variable (DRV) | Continuous Random Variable (CRV) |
| :--- | :--- | :--- |
| Values Obtained By | Counting (distinct, separate, countable values, e.g., number of students, coin flips) | Measuring (continuous scale along an interval, e.g., temperature, height, weight) |
| Core Properties | • Each individual probability lies between $0$ and $1$ <br> $0 \le P(x) \le 1$.<br>• The sum of all probabilities equals $1$ <br> $\sum P(x) = 1$. | • Probability at an exact single point is zero <br> $P(X = x) = 0$ <br>• Probability is measured as the *area under the curve* over a given interval. |
| Reliance | Probability Mass Function (PMF) to calculate the exact probability of specific, separate outcomes. | Probability Density Function (PDF) and integration, where probabilities are found over intervals rather than at exact points. |
| Expected Value $E(X)$ | • The long-term average or mean ($\mu$).<br>• Calculated by summing the products of each value and its probability:<br>  $E(X) = \mu = \sum (x \cdot P(x))$ | • Calculated using integration over the range of the continuous variable:<br>  $E(X) = \int_{-\infty}^{\infty} x \cdot f(x) \, dx$ |
| Variance ($\sigma^2$) & Standard Deviation ($\sigma$) | • $\sigma^2 = \sum \left((x - \mu)^2 \cdot P(x)\right)$ <br>• Standard Deviation: $\sigma = \sqrt{\sigma^2}$ | • Variance: $\sigma^2 = \int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx$<br>• $\sigma = \sqrt{\sigma^2}$ |
| Common Examples | Binomial, Poisson, Geometric, Hypergeometric | Uniform, Normal (Gaussian), Exponential, Student's t-distributions. |

## Discrete Probability Distribution Functions
Common discrete probability distributions include the **binomial**, **geometric**, **hypergeometric**, and **Poisson** distributions.

### Binomial Discrete Probability Distribution

**Characteristics of a Binomial Experiment:**
* Fixed number of independent trials, denoted by $n$.
* Exactly two mutually exclusive possible outcomes per trial: **success** ($p$) and **failure** ($q$), where $p + q = 1$.
* The probability of success ($p$) remains constant across every trial.

**Shortcut Formulas for a Binomial Experiment:**

| Parameter | Description | Formula |
| :--- | :--- | :--- |
| Mean ($\mu$) | Expected number of successes in $n$ trials. | $\mu = n \cdot p$ |
| Variance ($\sigma^2$) | Spread of the binomial distribution. | $\sigma^2 = n \cdot p \cdot q$ |
| Standard Deviation ($\sigma$) | Square root of the variance. | $\sigma = \sqrt{(n \cdot p \cdot q)}$ |
| Notation | Indicates $X$ is binomially distributed. | $X \sim B(n, p)$ |

## Continuous Probability Distribution Functions

### Uniform Distribution
* Concerned with continuous events that are equally likely to occur across a continuous interval from $a$ to $b$.

**Shortcuts of a Uniform Distribution:**

| Parameter / Feature | Description | Formula / Notation |
| :--- | :--- | :--- |
| Notation | $X$ is uniformly distributed with minimum $a$ and maximum $b$. | $X \sim \text{Uniform}(a, b)$ |
| Probability Density Function | Constant height of the probability density curve. | $f(x) = \frac{1}{b - a}$ for $a \le x \le b$ |
| Mean ($\mu$) | Midpoint of the interval. | $\mu = \frac{a + b}{2}$ |
| Variance ($\sigma^2$) | Spread based on the total width of the interval. | $\sigma^2 = \frac{(b - a)^2}{12}$ |
| Standard Deviation ($\sigma$) | Square root of the variance. | $\sigma = \sqrt{\frac{(b - a)^2}{12}}$ |

### The Standard Normal Distribution and Student's t-Distribution

* **Core Similarities:** 
  * Both distributions have continuous, symmetrical, and bell-shaped graphs.
  * Both assume that the underlying population data (or the sampling distribution of the mean via the Central Limit Theorem) is approximately normal.

* **Parameters of the Population from which the sample is drawn**
  
| Property / Feature | Standard Normal Distribution ($Z$) | Student's t-Distribution ($t$) |
| :--- | :--- | :--- |
| Mean $\mu$ | Unknown; Inferred from the sample mean $\bar {x}$ | Unkown; Inferred from the sample mean $\bar {x}$ |
| Standard Deviation $\sigma $| Known | Unknown; inferred from the sample standard deviation $s$ |

* **Statistics of the sample in the distribution**

| Property / Feature | Standard Normal Distribution ($Z$) | Student's t-Distribution ($t$) |
| :--- | :--- | :--- |
| Transformed for comparision across different datasets<br> $\mu$ is  unknown but used as a reference for formulae  to measure distance from. | **Z-score:** Number of standard deviations a value is from the mean.<br> Raw_score transformed to $z = \frac{x - \mu}{\sigma}$ <br> Sample Mean transformed to $z = \frac{\bar{x} - \mu}{\frac{\sigma }{\sqrt{n}}}$ | **t-score:** Number of standard errors the sample mean is from the population mean.<br> Sample Mean transformed to: $t=\frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}$<br> No transformation of row values as there is no population standard deviation |
| Mean $\bar {x}$ | Transformed to: $z$=$0$ <br> Graph symmetrical to the transformed mean| Transformed to $t$=$0$<br> Graph symmetrical to the transformed mean |
| Standard Deviation $s$ | Standardized to $z$=$1$| varies with degrees of freedom<br> $\sqrt (\frac {df}{df-2} )$ |
| Tail Thickness & Probability | Thinner tails due to lower probability of extreme values. | Thicker tails due to higher probability of extreme values.<br>Gradually approaches the normal distribution as degrees of freedom ($df = n - 1$) increase. |

**Empirical Rule (68-95-99.7 Rule) for the Standard Normal Distribution**
For any normally distributed random variable:

| Percentage of Data | Statistical Interval | Description |
| :--- | :--- | :--- |
| **68%** | $\mu \pm 1\sigma$ | Lies within one standard deviation of the mean. |
| **95%** | $\mu \pm 2\sigma$ | Lies within two standard deviations of the mean. |
| **99.7%** | $\mu \pm 3\sigma$ | Lies within three standard deviations of the mean. |

#### Single Population Mean and Confidence Intervals 

| Key term | Description |
| :--- | :--- |
| Single Population Mean | Estimating the population mean |
| Point estimate |a single sample statistic used to estimate an unknown population parameter (e.g sample mean $\bar {x}$ used to infer population mean $\mu$). |
| Interval estimate |a range of plausible values calculated from sample data that is likely to contain the true population parameter. |
| Confidence Level (CL) | Probability that the calculate interval estimate will contain the true population parameter<br> Researchers choose confidence intervals of *90% or higher* to be reasonably certain of their conclusions |
| Alpha ($\alpha$) | Probability that the confidence interval does *not* contain the unknown population parameter <br> $\alpha = 1 - \text{Confidence Level}$ |

**Calculating the Confidence Interval**
* Used samples to estimate where the true population mean lies .
* Applies  Normal or Student's t distribution depending on absence/presence of population standard deviation $\sigma $ .

| Distribution | Normal | Student's t-test  |
| :--- | :--- | :--- |
| Traditional usage |  For large samples ($n \ge 30$) | For small samples ($n < 30$) |
| Current usage | Population standard deviation $\sigma $ is known | Population standard deviation $\sigma $ is unknown <br> Sample standard deviation ($s$) is used as an estimate for the unknown population standard deviation ($\sigma$) |

* *Steps to calculate the confidence level of Y%*
    *  The central $y\%$ of the probability distribution must be included.
    * A total probability of $\alpha = 1 - y\%$ must be excluded.
    * Since $\alpha$ is split between the two tails, it is divided by $2$ to get the exact portion of each tail ($\alpha/2$).

| Known Population standard deviation $\sigma $(Normal) | UnKnown Population standard deviation $\sigma $(Student's t) |
| :--- | :--- |
| The critical Z-score corresponding to $\alpha/2$ is looked up in the Z-table.| The critical $t$-score corresponding to $\alpha/2$ and $df = n - 1$ is looked up in the $t$-table. <br>The table gives critical $t$-values that correspond to the confidence level (column) and degrees of freedom (row). |
| Error Bound for the Population Mean (EBM): <br> $EBM = z_{\alpha/2} \left( \frac{\sigma}{\sqrt{n}} \right)$| Error Bound for the Population Mean (EBM): <br> $EBM = t_{\alpha/2} \left( \frac{s}{\sqrt{n}} \right)$ |

* Confidence Interval (CI)
   * The final interval estimate is given by:
   * $\text{CI} = (\bar{x} - EBM, \; \bar{x} + EBM)$

* Interpretation of a Confidence Interval
* "$Y$ percent confident that the true population mean $\mu$ lies between value $A$ and value $B$."*

#### Hypothesis Testing
* Another way of making statistical inferences other than calculating Confidence Intervals.
  
**Assumptions for making Hypothesis tests**

| Perform hypothesis test of | Description |
| :--- | :--- |
| Single population mean μ using a Student's t-distribution | • Data should be a simple random sample. <br>• Sample should be from a population that is approximately normally distributed.<br>• sample standard deviation used to approximate the population standard deviation.<br> • t-test works even when the population is *not approximately normally distributed provided the sample size is large* |
| single population mean μ using a normal distribution (z-test) | • Data should be a simple random sample.<br> • Population should be normally distributed or your sample size is sufficiently large.<br>• The  population standard deviation is known |
| single population proportion $p$|• Conditions for the Binomial distribution must be meant;<br> 1. certain number $n$ of independent trials<br> 2. outcomes of any trial are success or failure <br> 3. each trial has the same probability of a success $p$ <br> • Shape of the binomial distribution must be similar to the shape of the normal distribution; ensured by having quantities $np$ & $nq$ must be greater than 5. <br> • All the above  ensure that the estimated binomial distribution of a sample approximated by the normal distribution with mean $\mu $ =$p$ and $\sigma $=$\sqrt (\frac {p \cdot q}{n})$ |

**Overview of hypotheses**

| Null Hypothesis($Ho$) | Alternative Hypothesis($Ha$) |
| :--- | :--- |
| No difference between a sample mean or proportion and a population mean or proportion. | claim about the population that is contradictory to $Ho$  |

* Researcher determines which hypothesis is supported by the sample and decides;

| Sample information |  Decision |
| :--- | :--- |
| favors alternative hypothesis | Reject the Null hypothes $Ho$ |
| Insufficient to reject the null hypothesis | Do not reject the Null Hypothesis $Ha$ |
  
**Mathematical Symbols Used in Null Hypothesis($Ho$) & Alternative Hypothesis($Ha$)**
* The *Null hypothesis $Ho$* always has an equal (`=`) sign in it.
* The *Alternative hypothesis $Ha$* commonly uses greater than or less than signs (`><`)

| Null hypothesis $Ho$ | Alternative hypothesis $Ha$ |
| :--- | :--- |
| equal (=)	| not equal (≠) or greater than (>) or less than (<) |
| greater than or equal to (≥) |	less than (<) |
| less than or equal to (≤) |	more than (>) |

**Outcomes and the Type I and Type II Errors**
* There are four possible outcomes of hypothesis testing, these outcomes depend on;
    1. Actual truthfulness or falseness
    2. Decision to reject or accept the Null Hypothesis $Ho$

| ACTION	| $Ho$ IS ACTUALLY	...| |
| :--- | :--- | :--- |
| | True | False |
| Do not reject $Ho$ | Correct outcome | Type II error |
| Reject $Ho$ | Type I error | Correct outcome |

* TYpe of Error

| Type | Description |
| :--- | :--- |
| $\alpha$ | probability of a Type I error<br> probability of rejecting the null hypothesis when the null hypothesis is true. |
| $\beta$| probability of a Type II error<br>probability of not rejecting the null hypothesis when the null hypothesis is false. |

* α and β should be as small as possible because they are probabilities of errors(rarely zero).
* *Power of the Test:* 
    * Calculated as `1- β`,
    * Should be as close to one(1) as possible.
    * Increased by increasing the sample size.

**Example of Steps for Hypothesis testing**

| Step | Description |
| :--- | :--- |
| Problem | • A factory claims their lightbulbs last an average of 1000 hours<br> • There is suspicion that this is false |
| Write the hypotheses | • $Ho$: $\mu $=1000 <br> •  $Ha$: $\mu $< 1000 |
| Choose a significance level| •Since the $Ha$ is one tailed, \alpha will be in one tail. <br> • Taking a significance level of 95% (0.95) means $\alpha$ = 1-0.95=0.05|
| Take a random sample | • A sample of size $n$=25 with a mean $\bar {x}$ =980 and standard deviation $s$=40 |
| Choose the distribution | • Since the population standard deviation  $\sigma$ is *unknown* ,it is approximated by the sample standard deviation $s$ using the student's t-test   |
| Calculate the t-test |  • t-test = $\frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}$ <br>• t-test=[(980-1000)/(40/sqrt(25)] =-2.5 |
| Lookup | Lookup the critical t-value for $df$=25-1=24, and $\alpha$=0.05 which translates to 1.711<br> Since the $Ha$ is left sided(less than),we take only the lowerboundary  |
| Decision | • Because the calculated test statistic of -2.5 is more extreme than the critical threshold of -1.711, it falls into the rejection region, leading us to reject the null hypothesis.  |
