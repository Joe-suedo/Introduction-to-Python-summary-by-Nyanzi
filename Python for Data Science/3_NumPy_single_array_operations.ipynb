## Array Operations

### Single array operations
*mathematical and trigonometric ufuncs*
* `math module` can only perform math and trig functions on scalars (Not arrays)
* sin,cos and tan expect `radians` not degrees
* arcsin,arccos and arctan also output `radians`

| operation | description & behavior | syntax |
| :--- | :--- | :--- |
| sin | computes the trigonometric sine element-wise for all values in the array.| np.sin(arr) |
| cos | computes the trigonometric cosine element-wise for all values in the array . | np.cos(arr)|
| tan | computes the trigonometric tangent element-wise for all values in the array | np.tan(arr) |
| arcsin | computes the inverse sine element-wise. | np.arcsin(arr) |
| arccos | computes the inverse cosine element-wise. | np.arccos(arr) |
| arctan | computes the inverse tangent element-wise. | np.arctan(arr) |
| radians or np.deg2rad | converts angles from degrees to radians | np.radians(arr) or np.deg2rad(arr) |
| degrees or np.rad2deg | converts angles from radians to degrees | np.degrees(arr) or np.rad2deg`|


```python
# Create array of angles
arr1=np.array([-90,-60,-30,0,30,60,90])
print(f'arr1 is an array of angles\n{arr1}\n')

#conversion of arr1 elements into radians
arr1_radians=np.radians(arr1)
print(f'np.radians(arr1) converts angles in arr1 to radians\n{arr1_radians}\n')

#Computing sin arr1
arr1_sin =np.sin(arr1_radians)
print(f'np.sin(arr1_radians) computes the sin\n{arr1_sin}\n')

#Computing the sin inverse 
arr1_sin_inv =np.arcsin(arr1_sin)
print(f'np.arcsin(arr1_sin) computes the sin inverse\n{arr1_sin_inv}\n')

# Converting the radians above back to degrees
arr1_degrees=np.degrees(arr1_sin_inv)
print(f'np.degrees(arr1_sin_inv) computes the original degrees\n{arr1_degrees}')
```

**Other common Single array functions**
* Reduction operators such as sum and prod process all elements across the *entire array by default* to return a single scalar
  
* Cumulative operations such as cumsum and cumprod compute running totals or products across the elements (flattening the array into a one-dimensional sequence if no axis is specified)
  
* Optional axis parameter (0 for rows and 1 for columns) can be provided to target a specific dimension 
* axis=0 collapses axis 0 (rows) down to a single summary, giving a result grouped by columns.
* axis=1 collapses axis 1 (columns) across, giving you a result grouped by rows.
  
| operation | description | syntax |
| :--- | :--- | :--- |
| sum | computes the sum of all elements in the array by default, or along a specified axis | arr.sum() |
| cumsum | computes the cumulative running sum of elements, flattening by default or operating along a specified axis | arr.cumsum() |
| prod | computes the product of all elements in the array by default, or along a specified axis | arr.prod() |
| cumprod | computes the cumulative running product of elements, flattening by default or operating along a specified axis | arr.cumprod() |
| min | computes the minimum value across all elements in the array by default, or along a specified axis | arr.min() |
| argmin | returns the indices of the minimum values along a specified axis by default, or flattening if no axis is specified | arr.argmin() |
| max | computes the maximum value across all elements in the array by default, or along a specified axis | arr.max()|
| argmax | returns the indices of the maximum values along a specified axis by default, or flattening if no axis is specified | arr.argmax() |
| ptp | computes the peak-to-peak range (maximum minus minimum) across the array by default, or along a specified axis | arr.ptp() |
| mean | computes the arithmetic mean of all elements in the array by default, or along a specified axis | arr.mean()|
| std | computes the standard deviation of all elements in the array by default, or along a specified axis | arr.std()|
| var | computes the variance of all elements in the array by default, or along a specified axis | arr.var() |
| sort | sorts the array in-place along a specified axis by default (typically the last axis) | arr.sort() |
| argsort | returns the indices that would sort the array along a specified axis | arr.argsort() |
| diagonal | exclusively extracts diagonal elements from existing arrays. | np.diagonal() (or arr.diagonal()) |

```python
# Create a multi dimensional array
arr3=np.array([[1,1,1,1,1],[1,2,3,4,5],[2,2,2,2,2]])
print(f'arr3 is a multidimensional array\n{arr3}\n')

#Compute the sums
arr3_sum=np.sum(arr3)
print(f'the sum of all elements in arr3 is {arr3_sum}\n')

arr3_sum_rows=np.sum(arr3, axis=1)
print(f'np.sum(arr3, axis=1) computes sum across rows {arr3_sum_rows}')

arr3_sum_columns=np.sum(arr3, axis=0)
print(f'np.sum(arr3, axis=0) computes sum across columns {arr3_sum_columns}')

arr3_min=np.min(arr3)
print(f'np.min(arr3) returns the minimum element in arr3 {arr3_min}')
```
