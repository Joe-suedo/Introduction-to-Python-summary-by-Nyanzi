# Descriptive statistics
* Organizing and summarizing data.
* This is by graphing and summarizing with numbers like average.
* The goal of statistics is  to gain an understanding of data (not to perform numerous calculations using the formulas)
  
**Inferential statistics**
 * For drawing conclusions from data.
 * Use probability to determine the confidence the drawn conclusions.

**Statistical Models**
 * Statistics uses mathematical models to describe phenomena that occur in the real world.
 * *Deterministic models*
     * Used when one value is precisely predicted from another value.
     * Quite accurate & used to make predictions with a high degree of precision.
     * Examples;
         * Quadratic equations that describe the acceleration of a car from rest
         * Differential equations that describe the transfer of heat from a stove to a pot. 

**Probability**
* Used to study randomness & deals with the chance of an event occurring
* Outcomes of a few repetitions are uncertain, but there is a regular pattern of outcomes when there are many repetitions.

## Qualitative Data

**Display of Qualitative Data**

| Method / Graph | Core Definition & Formula | Primary Application & Best Used When |
| :--- | :--- | :--- |
| Frequency & Relative Frequency| • Frequency ($f$): Raw count of observations per category.<br>• Relative Frequency: $\frac{\text{Frequency}}{\text{Total Observations}}$. | Summarizing qualitative datasets to reveal basic category distributions before graphing. |
| Bar Chart (Simple / Clustered) | • Uses rectangular bars where the length/height represents category frequency or percentage.<br>• Can be oriented horizontally or vertically. | Comparing discrete categories, tracking changes across groups,mapping cross-classified relationships using clustered bars or when some data is missing |
| Pie Chart / Sector Diagram | • Slices of a circle representing parts of a whole.<br>• Sector angle formula: $\text{Percentage} \times 3.6^\circ$. | Displaying proportional breakdowns (percentages adding up to 100%) when there is no missing data. |
| Pareto Chart | • A specialized bar chart where categories are sorted in descending order of frequency (largest to smallest). | Identifying the most frequent categories or vital few contributors in quality control and data analysis. |
| Pictogram / Map Diagram | • Pictogram: Uses icons or symbols scaled to represent data frequencies.<br>• Map Diagram: Depicts geographical or spatial distributions of qualitative traits. | Communicating categorical breakdowns to broad audiences visually or analyzing spatial-geographic disease/demographic trends. |

**Two-Way Frequency Tables and Marginal Distributions**

* *Bivariate Categorical Data:* Used when two qualitative (categorical) variables are measured simultaneously.
* *Contingency Table (Two-Way Table):* Data arranged in rows for one categorical variable and columns for the other.
* *Marginal Frequencies:* Found in the margins (the final "Total" row and final "Total" column) of a two-way table.Represent the overall distribution of one variable while ignoring the levels of the second variable.
* *Mathematical Expression:* Marginal probabilities/distributions can be expressed as raw joint frequency totals divided by the grand total ($N$):
* $\text{Marginal Relative Frequency} = \frac{\text{Row Total or Column Total}}{\text{Grand Total (N)}}$

*Example: Contingency Table of Preferred Sport by Sex*

* Row Variable: Sex (Boys, Girls)
* Column Variable: Preferred Sport (Football, Netball)

| Sex / Preferred Sport | Football | Netball | Row Total (Marginal Frequency) |
| :--- | :---: | :---: | :---: |
| Boys | 30 | 2 | *32* |
| Girls | 7 | 24 | *31* |
| **Column Total (Marginal Frequency)** | *37* | *26* | *63 (Grand Total)* |

*Marginal Distributions Derived from Table:*
* Marginal Distribution of Sex:
  * $\text{Boys} = \frac{32}{63} \approx 50.79\%$
  * $\text{Girls} = \frac{31}{63} \approx 49.21\%$
* *Marginal Distribution of Preferred Sport:*
  * $\text{Football} = \frac{37}{63} \approx 58.73\%$
  * $\text{Netball} = \frac{26}{63} \approx 41.27\%$

## Sampling Techniques, Errors, and Statistical Principles

* **Purpose:** Sampling is conducted to ensure the sample is representative of the population.
* **Probability Sampling Principle:** Ideally, every item or member of the population has a known and equal chance of being selected for the sample.

#### Comprehensive Overview of Sampling Methods

| Sampling Method | Description & Procedure | Advantages & Disadvantages |
| :--- | :--- | :--- |
| Simple Random Sampling | Every member of the population has an equal and independent chance of selection | Highly objective and unbiased, but requires a complete, up-to-date sampling frame. |
| Stratified Sampling | The population is divided into homogenous subgroups (strata). <br>Random samples are drawn proportionally or equally from each stratum. | Ensures representation of minority subgroups, but requires prior knowledge of population characteristics. |
| Cluster Sampling | The population is divided into naturally occurring groups (clusters). <br>Entire clusters are randomly selected for inclusion. | Cost-effective and efficient for geographically dispersed populations, but increases sampling error. |
| Systematic Sampling | Members of the population are listed in sequence.<br>Every $n^{\text{th}}$ individual is chosen after a random start. | Easy to implement and ensures even spatial/temporal distribution, but vulnerable to periodic cyclical bias. |
| Convenience Sampling | Uses readily available, accessible items.| Fast and inexpensive, but results are prone to high selection bias and lack generalizability. |

#### Sampling Errors, Non-Sampling Errors, and Practical Guidelines

| Term | Description |
| :--- | :--- |
| Sampling Errors| Discrepancies between sample statistics and population parameters caused by the natural variation of selecting a subset rather than the whole population<br> Often magnified by a small sample size.
| Non-Sampling Errors | Errors arising from factors unrelated to the sampling procedure itself, such as measurement mistakes, data entry errors, non-response bias, or using a defective counting device. |
| Practical Reality | A sample is rarely completely identical to the true population parameters due to inherent sampling variability. |
| Rule of Thumb |As a general statistical principle, the larger the sample size, the smaller the sampling error and the greater the estimate precision. |

#### Evaluation of statistical studies before accepting results

| What to Evaluate | Description |
| :--- | :--- |
| Samples |Must be representative of the population.<br> Biased samples are not representative of the population & give inaccurate and unreliable results.|
| Sample size(number of observations) |  Larger samples are better and too small samples are unreliable<br> Small samples maybe unavoidable & can still be used to draw conclusions |
| Casuality | Relationship between two variables does not mean that one causes the other.<br>They may be related (correlated) because of their relationship through a different variable. |

#### Unreliable/biased samples maybe caused by

| Cause | Description |
| :--- | :--- |
| Self selected samples | E.g Responses only by people who choose to respond, such as internet surveys |
| Undue influence | Data collection in a way that influences the response |
| Non-response or refusal of subject to participate |  Often, people with strong positive or negative opinions may answer surveys, which can affect the results. |
| Self-funded or self-interest studies | A study performed by a person or organization in order to support their claim. |
