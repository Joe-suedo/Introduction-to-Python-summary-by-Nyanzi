## Stem-and-Leaf Plots (Stemplots)

* Statistical tool used in EDA to organize and display quantitative data.
* Particularly effective for small- to moderate-sized datasets.
  
* Every data value is split into two parts:
  * Stem:Consists of the leading digit(s) of the data point.
  * Leaf:Consists of the final significant digit of the data point.
  * Example of Data Breakdown

| Original Observation | Stem | Leaf |
| :--- | :--- | :--- |
| 23 | 2 | 3 |
| 432 | 43 | 2 |
| 5432 | 543 | 2 |
| 9.3 | 9 | 3 |

**Construction Rules**

* Stems are listed vertically in a column from smallest to largest.
* Corresponding leaves are written horizontally to the right of each stem, separated by a vertical line.
    * Leaves sharing the same stem must be arranged in increasing order from left to right.
    * A space or comma is typically placed between individual leaves for clarity.
* There should be a key or legend to define the scale and units (e.g., specifying that $43 \mid 2$ represents $432$).

**Key Benefits & Interpretatio**
* Stemplots provide a quick graphical summary of a dataset while preserving the original numerical values.
* Allow easy visualization of the distribution shape, overall pattern, and data clustering.
* Allow easy spotting of outliers.
    * Outlier: observation of data that does not fit the rest of the data. It is sometimes called an extreme value.

## Side-by-Side Stem-and-Leaf Plots

* A graphical tool used to compare two separate datasets sharing a common range of values.
* Features a single, central column of stems, with the leaves for the first dataset extending to the left and the leaves for the second dataset extending to the right.
    * Left Leaves (Dataset 1):Leaf values increase in order moving outward from the stem (from right to left).
    * Right Leaves (Dataset 2):Leaf values increase in order moving outward from the stem (from left to right).

## Line Graphs

* The horizontal axis ($x$-axis) represents data values or time intervals.
* The vertical axis ($y$-axis) represents frequencies, counts, or measurements.
* Data points corresponding to the observations are plotted on the grid and connected sequentially using straight line segments to show trends over time or sequence.

## Bar Graphs

* Consist of discrete bars separated by spaces, indicating that the data is categorical rather than continuous.
* Bars can be arranged vertically(column charts) or horizontally, depending on the layout preference and the length of the category labels.

## Histograms

* Composed of contiguous (adjoining) bars that display the distribution of continuous quantitative data. 
* Unlike bar charts, the lack of spaces between bars emphasizes the continuous nature of the scale.
  
**Axes Definition:**
  * Horizontal ($x$-axis):Represents the *quantitative scale* of the variable, functioning as a continuous number line.
  
  * Vertical ($y$-axis): 
      * Represents frequency, relative frequency, percent frequency, or probability. 
      * The choice of vertical scale changes the numerical values on the axis but leaves the overall shape of the distribution unchanged.
  
        | Vertical Label | Description |
        | :--- | :--- |
        | Frequency ($f$)| The raw count of observations falling within a specific class interval. |
        | Relative Frequency ($\text{RF}$) | The frequency of a class divided by the total number of observations ($n$). |
        | Percent Frequency | The relative frequency expressed as a percentage ($\text{RF} \times 100\%$). |

**Construction Guidelines**

*Determine Class Intervals:*
  * Choose the number of bins (classes) to effectively summarize the data.
  * While there is no single rigid formula, standard guidelines exist:
    * Sturges' rule
      * Base 10 version: $k = 1 + 3.322 \log_{10}(n)$
      * Base 2 version: $k = 1 + \log_2(n)$
    * Square-root choice:
      * $k = \sqrt{n}$ (often rounded up to the next whole number)
  
*Calculate Bin Width:*
* Estimate the width of each bar by dividing the range of the data values by the chosen number of bins:

$\text{Bin Width} \approx \frac{\text{Maximum-Minimum}}{\text{Number of Bins}}$

**Statistical Interpretation**

Histograms provide a comprehensive visual summary of a dataset, highlighting three core properties:
* Shape: Shows the distribution pattern of the data—whether it is symmetric, approximately normal, uniform, or skewed in a particular direction.
* Center: Highlights the approximate middle or typical value of the dataset.
* Spread: Indicates the variability, showing how widely the data values are dispersed around the center.

## Frequency Polygons

* For understanding the shapes of distributions, analogous to line graphs, created by connecting the midpoints of the tops of the bars in a histogram.
* Particularly useful for comparing two or more distributions because multiple frequency polygons can be plotted on the same set of axes.

**Construction**

*  Examine the data and decide on طبthe number of intervals and resulting interval size, for both the $x$-axis and $y$-axis.
*  $x$-axis: Shows the scale of measurement, typically labeled with the *class midpoints* (the halfway point between the lower and upper bounds of each interval) rather than just the boundaries, to anchor the line segments.
* $y$-axis: Represents the frequencies, relative frequencies, or percentages of the values.
* Each data point represents the frequency for each interval, plotted at the midpoint (halfway between the lower and upper bound).
* The polygon is typically "closed" at both ends by extending the line down to the $x$-axis at the midpoint of an imaginary interval before the first class and after the last class, ensuring the graph forms a complete enclosed area.
  
## Time Series Graphs

* A specialized graphical format used to display data collected over sequential time intervals, making long-term movements and trends easy to spot.
* Horizontal ($x$-axis): Used to plot the date or time increments (such as days, months, or years).
* Vertical ($y$-axis): Used to plot the values of the variable that we are measuring.
* The points on the graph are typically connected by straight lines in the order in which they occur.

### Summary of Graphical Methods & Use Cases

| Graph / Display Type | Primary Use Case / When to Use | Key Differentiator (What Sets It Apart) |
| :--- | :--- | :--- |
| Stem-and-Leaf Plot (Stemplot) | Organizing small- to moderate-sized quantitative datasets in Exploratory Data Analysis (EDA). | Preserves exact original values: individual data points remain visible by splitting numbers into stems and leaves. |
| Side-by-Side Stem-and-Leaf Plot | Comparing two separate datasets sharing a common range of values. | Dual-sided layout: Uses a single central stem column with leaves extending in opposite directions (left vs. right) with inverse ordering rules. |
| Bar Graphs | Displaying and comparing categorical data. | Categorical & Separated: Features discrete, separated bars, emphasizing distinct groups rather than continuous scales. |
| Histograms | Displaying the distribution of continuous quantitative data. | Contiguous bars & Bins: Uses adjoining bars (no spaces) to emphasize continuity, highlighting the shape, center, and spread of the data. |
| Line Graphs | Showing trends over sequential data or observations over time. | Sequential connection: Plots individual data points and connects them sequentially with straight line segments. |
| Frequency Polygons | Visualizing and comparing distribution shapes. | Midpoint-anchored & Overlay-friendly: Built by connecting the interval midpoints of a histogram, making it the preferred choice for overlaying and comparing *multiple* distributions on the same axes. |
| Time Series Graphs | Displaying data collected over sequential time intervals. | Timee-specific $x$-axis: A specialized line graph where the horizontal axis is strictly restricted to time increments (dates, months, years) to track long-term trends. |

### Key Comparisons & Distinctions

#### 1. Line Graphs vs. Frequency Polygons
* *Line Graphs:* Focus on tracking individual observations or general sequential data points where the $x$-axis can represent arbitrary data values or sequential categories.
* *Frequency Polygons:* Specifically derived from continuous frequency distributions (histograms). They are anchored to *class midpoints* on the $x$-axis and are structurally "closed" at both ends to touch the $x$-axis, making them ideal for overlaying multiple distributions on a single plot.

#### 2. Bar Graphs vs. Histograms
* *Bar Graphs:* Used for *categorical data* (qualitative or discrete nominal/ordinal variables). Bars are separated by spaces to indicate distinct categories.
* *Histograms:* Used for *continuous quantitative data*. Bars touch one another (contiguous) to visually represent continuous ranges or intervals across a number line.

#### 3. Line Graphs vs. Time Series Graphs
* *Line Graphs:* Can show relationships or sequential trends across any ordered numeric variable.
* *Time Series Graphs:* A specialized subset of line graphs where the $x$-axis is exclusively dedicated to chronologically ordered time units (e.g., hours, days, months, years).
