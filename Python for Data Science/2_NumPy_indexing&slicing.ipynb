### Array Indexing And Slicing

**Indexing**
* Arrays can be indexd and are mutable

| Array Type | Description & Behavior | Syntax & Example |
| :--- | :--- | :--- |
| 1D Arrays | Uses zero-based indexing identical to standard Python lists.<br>Negative indices count backward from the end (-1 is the last element). | arr[index]<br>arr[0] *(first)*<br>arr[-1] *(last)* |
| Multi-Dimensional Arrays | Indexed using comma-separated integers (a tuple of indices) for each axis.<br>• For 2D arrays: [row_index, column_index] (Axis 0 = Rows, Axis 1 = Columns).<br>• Supports single-axis selection (e.g., specifying only the row returns the entire row). | arr[row, col]<br> arr[0, 2] *(row 0, col 2)*<br>arr[1] *(returns row 1 as 1D array)* |

```python
import numpy as np
```

```python
# 1D Arrays indexed like standard Python lists
arr_a=np.array(['Saka','Ben','Eze','Konsa','Myles','Noni'],dtype=str)
print(f"arr_a[0] extracts the first item from ['Saka','Ben','Eze','Konsa','Myles','Noni']: {arr_a[0]}")
print(f"arr_a[-1] extracts the last item from ['Saka','Ben','Eze','Konsa','Myles','Noni']: {arr_a[-1]}")
```

```python
# Indexing Multidimensional arrays 
arr_b=np.array([[1,2,3,4],[4,3,2,1],[5,6,7,8]])
print(f"{arr_b}\narr_b[0] extracts the first first row\n{arr_b[0]}\n")
print(f"arr_b[(0,0)] extracts the first element in the array\n{arr_b[(0,0)]}\n")
print(f"arr_b[(-1,-1)] extracts the last element in the array\n{arr_b[(-1,-1)]}\n")
print(f"arr_b[(0,0),(0,1)] extracts the first 2 elements in the first row of the array\n{arr_b[(0,0),(0,1)]}\n")
```

**Array Slicing**

| Category / Type | Description & Behavior | Syntax & Example |
| :--- | :--- | :--- |
| 1D Array | Uses `[start:stop:step]` notation with the *stop value is exclusive*.<br> Omitting bounds defaults to the start, end, and a step of `1`. | arr[start:stop:step]<br>arr[::-1] *(reverses the array)* |
| Multi-Dimensional | Extracts sub-matrices by applying slice ranges across multiple axes separated by a comma. | arr[row_start:row_end, col_start:col_end]|
| Row & Column Range Slicing| Using a standalone colon `:` across an axis selects the entire row or column range while slicing the other dimension. | arr[row_slice, :] or arr[:, col_slice] |
| Particular Row Selection | Selecting a single row index without specifying columns returns that specific row as a 1D array. | arr[row_index]<br>arr[1] extracts 2nd row(index 1) |
| Particular Column Selection | Selecting a specific column requires a colon `:` for rows to span all rows, paired with the target column index. | arr[:, col_index]<br>arr[:, 0] (extracts 1st column)|

```python
#Slicing 1D array
print(f'arr_a is {arr_a} ')
print(f'arr_a[0:5:2] extracts from the 1st element to the last(exclusive) in steps of 2 \n{arr_a[0:5:2]}')
print(f'arr_a[0::2] same as arr_a[0:5:2] \n{arr_a[0::2]}')
```

```python
# Slicing a multidimensional arrays
print(f'arr_b is\n{arr_b}\n')
print(f'arr_b[0,:] extracts the first column\n{arr_b[0,:]}\n')
print(f'arr_b[:,0] extracts the first row\n{arr_b[:,0]}\n')
print(f'arr_b[0,2:1,3] extracts elements in the 1st and 2nd row plus the 2nd & 3rd column\n{arr_b[0:2,1:3]}\n')
```

**Copying and Viewing Arrays**

| concept / method | description & memory behavior | syntax |
| :--- | :--- | :--- |
| array views (slicing) | Array slices in numpy create views rather than copies<br>Thus the sliced variable shares memory with the original array.<br>Modifications made to either affect both. | arr[start:stop] |
| np.may_share_memory() | Checks if two arrays share the same memory block and returns a boolean value (True or False). | np.may_share_memory(a, b) |
| np.copy() | creates a deep copy of an array in independent memory, ensuring modifications to the copy do not alter the original array. | np.copy(arr) |


```python
#Views 
print(f'arr_a is {arr_a} ')
#Slicing arr_a
arr_a_sliced=arr_a[::2]
print(f'arr_a[::2] is {arr_a[::2]}')
#A change in sliced arr_a reflects in the parent arr_a
arr_a_sliced[0]='Star'
print(f'Index 0  was changed in both sliced and parent array\nparent array{arr_a}\nsliced array{arr_a_sliced}')
#Arrays that share memory can be checked
print(f'arr_a and arr_a_sliced share the same memory: {np.may_share_memory(arr_a,arr_a_sliced)}')
```

```python
#Copys
#Arrays that dont share memory with their parents can be created with copy
arr_a_copy=np.copy(arr_a[::2])
# A modification in the copy or parent cannot  be reflected in the other
print(f'arr_a is {arr_a}')
print(f'arr_a_copy is {arr_a_copy}') 
arr_a_copy[1]='Ebz'
print(f'arr_a is now {arr_a}')
print(f'arr_a_copy is now {arr_a_copy}') 
#And these dont share memory
print(f'arr_a and arr_a_copy share the same memory: {np.may_share_memory(arr_a,arr_a_copy)}')
```

**Fancy Indexing**
* Unlike basic slicing, fancy indexing returns a **copy** of the data rather than a view.

| Category / Type | Description & Behavior | Syntax & Example |
| :--- | :--- | :--- |
| Integer Fancy Indexing | Passing lists or arrays of integers to select non-contiguous rows, columns, or specific elements in an arbitrary order.  | arr[[row1, row2]]<br>`arr[[0, 2], [1, 3]]` *(selects specific coordinate pairs (0,1) and (2,3))* |
| Boolean Fancy Indexing | Filters or selects elements using a boolean mask (an array of `True`/`False` values of matching shape) generated via conditional expressions. Multiple conditions can be combined using bitwise operators (`&` for AND, `\|` for OR, `~` for NOT) or or NumPy logical functions. | `arr[arr > 5]`<br>`arr[(arr > 2) & (arr < 8)]`|
| Conditional Selection (`np.where`) | Evaluates a boolean condition and selects elements from either `x` (if true) or `y` (if false). <br>Useful for conditional replacement or filtering without reshaping the output structure. | `np.where(condition, x, y)`<br>`np.where(arr > 5, arr, 0)` *(keeps value if > 5, else replaces with 0)* |

** NumPy Logical Functions**
* Used to generate boolean arrays for NumPy Boolean Indexing

| Function | Evaluates to `True` Elementwise If | Syntax & Example |
| :--- | :--- | :--- |
| np.logical_and() | * Both conditions are satisfied<br> * Alternative to the `&` bitwise operator | np.logical_and(arr > x, arr < y) |
| np.logical_not() | * The condition is not satisfied<br> * Alternative to the `~` operator  | np.logical_not(x)<br>np.logical_not(arr > 5) |
| np.logical_or() | * At least one condition is satisfied <br> * Alternative to the `\|` bitwise operator| np.logical_or(arr < 2, arr > 8) |
| np.logical_xor() | * Strictly one condition is true.<br> * False even when both conditions are `True`.<br> * Alternative to the `\^` bitwise operator|| np.logical_xor(arr > 2, arr > 6) |

```python
# Integer fancy Indexing
arr_c = np.array(['Arsenal','Spurs', 'Man Utd', 'Chelsea', 'Man City', 'Barca'])
arr_c1=arr_c[[1,3,5]]
print(arr_c1)
```

```python
# Boolean fancy Indexing
arr_d = np.array([[1,2,3,1],[2,3,8,7], [13,9,1,2],[5,4,4,4]])
print(f'arr_d is\n{arr_d}\n')

#Creation of a boolean array
arr_d_greater_or_equal3= (arr_d==3) | (arr_d>3)
print(arr_d_greater_or_equal3)
arr_d_slice = arr_d[arr_d_greater_or_equal3]
print('\n',arr_d_slice)
```
