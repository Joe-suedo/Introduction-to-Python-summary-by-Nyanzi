# NumPy MODULE

## Introduction to NumPy
* The NumPy package is the workhorse of data analysis, machine learning, and scientific computing in the python ecosystem.
* Some of the leading packages rely on NumPy e.g scikit-learn, SciPy, pandas, and tensorflow.

* *Core python limitations:* 
  * Standard python lacks a native, high-performance array structure. 
  * Although nested lists can simulate multidimensional arrays, they lack vectorization, memory optimization, and dedicated matrix math functions.

* *The NumPy (Numerical Python Solution)* 
  *  Foundational library for scientific computing in python, specifically built for manipulating large, multidimensional arrays and matrices efficiently.
  * Requires significantly less memory and provides vectorized execution close to hardware speeds.
  * As an external library, numpy must be installed via package managers (`pip` or `conda`) and imported into scripts using the universal alias:
  ```python
  import numpy as np
  

````python
import numpy as np
````

*1. Using NumPy's Built-in Functions*

| Function | Description & Boundary Behavior | Syntax |
| :--- | :--- | :--- |
| arange | Generates a 1D array of evenly spaced values exclusive of the stop value(like `range` )<br>Works by specifying the **exact step value** between the first and last value, by default this step is 1.<br>Operates like standard Python `range()` by excluding the stop value. | np.arange(start, stop, step) |
| linspace | Generates a 1D array with a specified number of evenly spaced points. The stop value is inclusive by default but can be made exclusive by setting `endpoint=False`<br> Works by specifying the **exact number of elements** between the first and last value<br>Setting `retstep=True` returns a tuple containing both the array and the calculated step size (the exact distance between consecutive points). | np.linspace(start, stop, num) |
| eye | Creates a 2D identity matrix with ones on the main diagonal and zeros elsewhere. | np.eye(m,n) |
| diag | Extracts a diagonal or constructs a diagonal array of dimensions (m,m)from a given 1D sequence. | np.diag(vector) |
| logspace | Generates a 1D array of numbers spaced evenly on a logarithmic scale.<br>The start and stop values represent the powers of the base (default base is 10).| np.logspace(start, stop, num) |

````python
# 1. Using NumPy's builtin functions
# array from 0 t0 10(exclusive) in steps of 2
arr1= np.arange(0,10,2) 
print(f'np.arange(0,10,2) creates an array with values 0 to 10 in steps of 2\n{arr1}\n')

# array of 5 equally spaced elements from 0 to 10(inclusive)
arr2 = np.linspace(0,10, 5, retstep=True) 
print(f' np.linspace(0,10, 5, retstep=True) creates an array of 5 equally-spaced values from 0 to 10(inclusive)\n{arr2}\n')

# array of 5 equally spaced elements from 0 to 10(exclusive)
arr3 = np.linspace(0,10, 5, endpoint=False, retstep=True) 
print(f'np.linspace(0,10, 5, endpoint=False, retstep=True) creates an array 5 equally spaced values from 0 to 10(exclusive)\n{arr3}\n')

# 3 by 3 array with a diagonal of 1s
arr4 = np.eye(3,3)
print(f'np.eye(3,3) creates a 3x3 array with (1,1,1) on the main diagonal & zeros elsewhere\n{arr4}\n')

# The diagonal can be extracted with diag
print(f'np.diag(arr4) extracts the diagonal of arr4 \n{np.diag(arr4)}\n')
````

**3. Array Datatypes**
* Using appropriate data types allows for compact, optimized data storage in memory.
* Accessed using the `.dtype` attribute (`array_variable.dtype`).
* Datatypes can be explicitly specified during array creation by passing the `dtype=` parameter (e.g float,int,bool, str).
* If the `dtype` argument is omitted during array creation, NumPy automatically infers the data type from the input data.

| Input Data | Inferred Datatype |
| :--- | :--- |
| Whole numbers | int (typically int64 or int32 depending on the system architecture) |
| Decimals / Floating points | float (typically float64) |
| Functions like np.zeros | float (default for initialization functions) |

*Comprehensive Datatype Reference*

| Datatype Category | Common Symbols / Aliases | Description & Range | Examples |
| :--- | :--- | :--- | :--- |
| Signed Integer | int64, int32, int16, int8 | Represents whole numbers, including positive, negative, and zero. | 4, -5, 100 |
| Unsigned Integer | uint64, uint32, uint16, uint8 | Represents *non-negative* whole numbers only (zero and positive)<br> Doubles the positive range limit by omitting negative signs. | 0, 255 (uint8) |
| Float (Decimal) | float64, float32, float16 | Numbers with fractional components containing decimal points. | 4.5, 3.14, 4. |
| Complex Number | complex128, complex64 | Numbers containing both a real and an imaginary component (denoted by `j`). | 3 + 5j, 1.5 - 2j |
| Boolean | bool | Logical data type that strictly evaluates to truth values. | True, False |
| String / Unicode | <U..., S... | Textual data. S denotes fixed-length ASCII bytes; <U denotes variable or fixed-length Unicode characters. | 'Nyanzi', 'Bioinformatics' |
| object | object, O | catch-all datatype that stores references (pointers) to python objects scattered across memory, enabling arrays of heterogeneous or mixed data types<br>Used to transform heterogeneous pandas Dataframes to numpy arrays to prevent errors during transformation | np.array(['text', 123, {'key': 'value'}, dtype=object]) |

*Note on Datatype Notations & Memory Suffixes*

* Numeric Bit-Width Suffixes (`32`, `64`, etc.):
  * Indicate the bit-width allocated per element in memory (int64 allocates 64 bits / 8 bytes per item).
  * This strictly dictates both numerical range limits and precision precision levels.
* Unsigned vs. Signed Range Difference:
  * An 8-bit signed integer (int8) spans from -128 to 127.
  * An 8-bit unsigned integer (uint8) spans from 0 to 255.
* String Prefixes (S, <U):
  * S: Fixed-length ASCII byte strings (e.g., `S7` allocates exactly 7 bytes per item).
  * <U: Unicode strings where the trailing number defines the maximum character capacity (e.g., `<U6` holds up to 6 characters).

````python
# Array datatypes
# dtype can be used to asign datatype during creation of arrays
arr16=np.array([True,False])
print(f'NumPy assigns {arr16.dtype} to the array created by np.array([True,False])\n')

arr17=np.array([True,False], dtype=int)
print(f'Since T or F can be converted to int,np.array([True,False], dtype=int) is of datatype {arr17.dtype}\n')

arr18=np.array([True,False], dtype=str)
print(f'Similary, np.array([True,False], dtype=str) is of datatype {arr18.dtype}\n')
````

**Array Size & Dimensions**

| Function / Attribute | Description & Behavior | Syntax & Example |

| shape | Returns a tuple representing the size of the array along each dimension `(rows, columns)`. <br>Can be accessed directly as an attribute or queried via the NumPy function. | arr.shape<br>np.shape(arr)|
| ndim| that returns an integer representing the total number of dimensions (axes) of the array.| arr.ndim|
| size | Returns the *total number of elements* across all dimensions combined.<br>Equivalent to $rows \times columns$). | arr.size |
| reshape() | Returns a view (sharing the underlying memory by default) with a modified shape, provided the total element count remains identical.<br>Modifying the reshaped array will affect the original array.<br>Can occasionally fall back to a copy if the array is non-contiguous. <br>Can be accessed directly as an attribute or queried via the NumPy function.| arr.reshape(m, n)<br>np.reshape(arr, (m, n)) |
| transpose / T | Reverses or permutes the axes of an array.<br> For a 2D matrix, it flips the axes by swapping rows and columns (Axis 0 becomes Axis 1).<br> Returns a view sharing memory. | arr.T<br>arr.transpose() |
