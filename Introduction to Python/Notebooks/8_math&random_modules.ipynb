# MATH AND RANDOM MODULES
## Math Module
* Houses most of the familiar mathematical functions. 
* `from` keyword :import only the specific methods you need from a library rather than loading the entire module. `from math import log, exp`
  
**Common Math Module Functions and Constants**

| Category / Item | Description | Example / Usage |
| :--- | :--- | :--- |
| Importing | Loads the math library into your script. | import math |
| Arithmetic | • math.sqrt(x): Returns the square root.<br>• math.factorial(x): Returns the factorial.<br>• math.gcd(a, b): Returns the greatest common divisor. | math.sqrt(25) <br>(Returns 5.0) |
| Rounding | • math.ceil(x): Rounds a number up to the nearest integer.<br>• math.floor(x): Rounds a number down to the nearest integer. | math.ceil(4.2) <br>(Returns 5) |
| Exponents & Logs | • math.pow(x, y): Returns x raised to power y.<br>• math.log(x): Returns the natural logarithm. | math.pow(2, 3) <br>(Returns 8.0) |
| Trigonometry | • math.sin(x), math.cos(x), math.tan(x): Trigonometric ratios (angles in radians).<br>• math.radians(x): Converts degrees to radians. | math.radians(180) <br>(Returns 3.14159...) |
| Constants | • math.pi: Mathematical constant pi.<br>• math.e: Mathematical constant e. | math.pi |


## Random module
  
| Determinist computer programs | Non determinist computer programs |
| :--- | :--- |
| Generate the same outputs all time given the same inputs | Generate the different outputs all time given the same inputs |

* Programs can be made nondeterminist by using alogorithms to generate pseudorandom numbers.
* Psuedorandom numbers are not random as they are generated by a deterministic computation.
**Random module** provides functions that generate pseudorandom numbers

**Common Functions of the Random module**

Common Random Module Functions

| Function / Syntax | Description | Example / Usage |
| :--- | :--- | :--- |
| import random | Loads the random module into your script. | import random |
| random.random() | Returns a random floating-point number between 0.0 and 1.0 (exclusive of 1.0). | random.random() <br>(Returns e.g., 0.3745) |
| random.randint(a, b) | Returns a random integer N such that a <= N <= b (inclusive). | random.randint(1, 10) <br>(Returns e.g., 7) |
| random.choice(sequence) | Returns a randomly selected element from a non-empty sequence (like a list, tuple, or string). | random.choice(['apple', 'banana', 'cherry']) <br>(Returns e.g., 'banana') |
| random.shuffle(x) | Randomly reorders the elements of a mutable sequence (like a list) in place. | numbers = [1, 2, 3, 4]<br>random.shuffle(numbers) |
| random.sample(population, k) | Returns a k-length list of unique elements chosen from the population sequence. | random.sample([10, 20, 30, 40], 2) <br>(Returns e.g., [30, 10]) |
| random.uniform(a, b) | Returns a random floating-point number between a and b. | random.uniform(1.5, 5.5) <br>(Returns e.g., 3.82) |
