## LISTS
* Store and organize multiple values in a single variable.
* One of the most useful data types in Python
* Different data types like floats, bool and strings can be stored together
* Element: item in the list
* Can be nested
  
**Creating Lists**

| Method | Syntax Example | Notes |
| :--- | :--- | :--- |
| Square Brackets (`[]`) | my_list = [4.0, 3, True, 'Joe']| The standard and common way  |
| `list()` Constructor | my_list = list((4.0, 3, True, 'Joe')) | Converts an iterable like a tuple into a list.Accepts *only one argument*, wrap multiple args in a double parantheses e.g.,list((4.0, 3, True))|
| split | str1.split() | Splits a string into a list of substrings based on a delimiter (defaults to whitespace).|

* `join` is the opposite of split; takes a list of strings and concatenates the elements into a string

**Properties of lists**

| Characteristic | Explanation | Example |
| :--- | :--- | :--- |
| Ordered | Can be indexed or sliced with a bracket operator housing an expression for the index or slice.| `my_list[0]` returns element at index 0 ,my_list[0:3] returns the first 3 elements |
| Mutable | Element at a particular index can be modified directly| `my_list[0] = 4` changes the first element in my list to 4 |
| Heterogenous | Each element can be a different type | [9, 9.0, 'nine', `True` ]|


**More on Indexing**
* Indexing:Accessing specific elements in a list by referring to their position
* Python indxing starts at zero(0)
* Any integer  can be used as an index.
* Index with a negative value: starts from the end of the list e.g., [-1] is the last element)
* `IndexError`: Raised on attempt to read or write an index position that does not exist
  
**More on slicing**
* Slicing:Extracts consecutive elements from a list.
* Syntax (`[start:stop]`):Extracts elements from the start index up to one index before the stop .
* Omitting the start (`[:stop]`):Extracts elements from the beginning(index 0) up to one index before the stop value.
* Omitting the stop (`[start:]`): Extracts elements starting from the specified index all the way to the end of the list.
* Can be used to update multiple elements
  
**Functions Commonly used with Lists**

| Function | Use |
| :--- | :--- |
| len | Returns the total number of elements in a list |
| in | Checks if an element exists in a list, returning a boolean (`True` or `False`) |
| for loop | Iterates through elements sequentially, Inner list in a nested list treated as one element  |

**List Operations**

| Operator | Description / Use |
| :--- | :--- |
| `+` (Concatenation) | Combines/concatenates elements from two or more lists into a single new list |
| `*` (Repetition) | Repeats a list a specified number of times |
| max | maximum value in the list |
| min | minimum value in the list |
| sum| sums all elements in a list|

**List methods**
* Most list methods are void; they modify the list and return None. 

| Method / Function | Description |
| :--- | :--- |
| .append() | Adds a single element to the end of a list. Common error: returns `None`, it only modifies the original list. |
| .insert(index, item) | Adds a new element at a specified index, shifting existing elements to the right. |
| .extend() | Appends all elements of another list into the current list (similar to using `+` or `+=`). |
| .sort() | Sorts a list *in-place* (modifies the original list directly and returns `None`)., Option `reverse=True` sorts according to len |
| sorted() | A built-in function that takes any iterable, sorts it, and *returns a new sorted list* without modifying the original. |
| .pop() | Removes and returns the element at a specified index (defaults to the last element if no index is provided). |
| .del statement | Deletes an element at a specified index or slice position (does not return or output the removed item). |
| .remove() | Removes the first occurrence of a specified element from the list (raises a `ValueError` if not present). |
| .reverse() | Reverses the elements of the list **in-place** (modifies the original list and returns `None`). |
| .index(item) | Returns the index of the first occurrence of item |
| .count(item) | Returns the number of occurrences of item |

**Alaising**
* Reference:  association of a variable with an object like a list
* Aliased object: Object can have more than one reference.(More than one variable name)
  * If the aliased object is mutable, changes made with one variable are carried on to the other:
  * This is error-prone, & generally safer to avoid aliasing when working with mutable objects.
  
**Debugging code**
* Careless use of lists (and other mutable objects) can lead to long hours of debugging. Ways to avoid these pitfalls.
  * Most list methods modify the argument and return `None`.Necessary to test code in iterative mode.
  * Too many ways to do the same thing, choose one method and stick with it.
  * Make copies to avoid alaising

**List Comprehension**
* Provides a compact, one-line syntax for creating new lists from existing iterables.
* Syntax:`[expression for item in iterable]`
* Key Notes:
  * Enclosed entirely in square brackets `[]`[cite: 8].
  * Omits the trailing colon (`:`) used in traditional `for` loops[cite: 8].

```python
# Functions used with lists
print('List a:', a)
print('Length of list a:', len(a))
print('List b:', b)
print('List b has element 3: ', 3 in b)
for element in b:
    print('Element:', element)  
```

```python
#LList methods
print('List a:', a)
a.append('New Element')
print('List a after append:', a)
a.insert(1, 'Inserted Element')
print('List a after insert:', a)
```

```python
#List comprehension
x = [1,2,3,4,5]
y = [i**2 for i in x]
print('List y:', y)
```
