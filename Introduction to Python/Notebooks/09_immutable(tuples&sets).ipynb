# TUPLES AND SETS

## TUPLES
*In short : tuples are immutable lists* 
* for holding data that must be write-protected
* No methods like append or pop; immutable
  
**Properties of Tuples**

| Characteristic | Explanation | Example |
| :--- | :--- | :--- |
| Ordered | Can be indexed or sliced with a bracket operator housing an expression for the index or slice.| `my_tuple[0]` returns element at index 0 ,my_list[0:3] returns the first 3 elements |
| Immutable | Element at a particular index *cannot* be modified directly| `my_tuple[0] = 4` raises an error |
| Heterogenous | Each element can be a different type | (9, 9.0, 'nine', `True` )|

**Syntax for creating a tuple:**

| Description| Syntax |
| :--- | :--- | 
| Empty tuple |`()` or `tuple()` |
| Tuple with one element | `(element,)`|

**Creating Tuples with Multiple Elements**

| Method / Approach | Description | Example & Behavior |
| :--- | :--- | :--- |
| Tuple Literal | Combines multiple comma-separated values directly inside parentheses. | my_tuple = (1, 2, 3) <br>(Creates a tuple successfully) |
| Constructor (Correct) | Passes a single iterable (such as a list) into the built-in tuple() function. | my_tuple = tuple([1, 2, 3]) <br>(Converts the list into a tuple) |
| Constructor (Incorrect) | Passing multiple separate arguments directly into the tuple() function. | tuple(1, 2, 3) <br>(Raises a TypeError because tuple() accepts at most one argument) |

**Methods for elements in tuples**

| Method | Description | Example |
| :--- | :--- | :--- |
| `index` method | finds the index of an element in a tuple | my_tuple.index(element) |
| `count` method | counts the number of times a particular element appears in a tuple | my_tuple.count(element) |
| `in` membership operator | Returns a boolean; `True` when element exists in the tuple or `False` otherwise | element in my_tuple |
| sorted() | Built-in function that returns a new sorted list from set items (requires comparable data types). | sorted(my_tuple) |


**Tuple unpacking**
* Extracts elements from a tuple and assigns them to multiple individual variables in a single statement.

| Feature / Scenario | Description | Example |
| :--- | :--- | :--- |
| Requirement | The number of variables on the left-hand side must match the exact number of elements in the tuple on the right-hand side | x, y, z = (1, 2, 3) |
| Error Handling | Raises a ValueError if the number of variables does not match the number of tuple elements. | x, y = (1, 2, 3) raises ValueError |
| Swapping Variables | Allows clean swapping of two variable values without needing a temporary third variable. | `a, b = b, a` means a is 1 and b is 2 |

## SETS
* Used to perform typical set operations such as union, intersection, set difference.
* Can be used to deduplicate lists
* Sets behave a lot like a dictionary that only contains keys (and no values). 
* Sets allow quick checking if an item is or not contained in them with the `hash` alogorithm.
* `Hash` : represents an immutable data type with a unique numerical representation

**Properties of Sets**

| Property | Description |
| :--- | :--- |
| Mutable | The set can be modified; items can be added or removed. |
| Unordered | Elements do not maintain a fixed sequence or position; cannot be indexed or sliced. |
| Unique Elements | Sets automatically remove duplicates as every element must be distinct. |
| Immutable Elements | Elements cannot be modified after creation.<br>Immutable data types like integers, floats, and strings plus immutable data structures like tuples are allowed.<br>Mutable data structures like lists and dictionaries are prohibited. |
| Hashable | Elements have a fixed, unchangeable hash value for checking its presence or absence in a set.<br>Only immutable objects have a hash value.<br>`hash(x)` gives the hash value of an item and raises an error is the item is unhashable |

**Creation of sets**

Creating Sets in Python

| Method / Approach | Description | Example |
| :--- | :--- | :--- |
| set() constructor | Creates an empty set. Using empty curly braces `{}` initializes an empty dictionary instead. | my_set = set() |
| Set literal (`{...}`) | Creates a set populated with initial elements.| my_set = {element1, element2} |

**Set methods**

| Method / Function | Description | Example |
| :--- | :--- | :--- |
| sorted() | Built-in function that returns a new sorted list from set items (requires comparable data types). | sorted(my_set) |
| .add() | Adds a single element to the set. | my_set.add(item) |
| .discard() | Removes an element from the set without raising an error if the item is absent. | my_set.discard(item) |
| .remove() | Removes an element from the set; raises a KeyError if the item is absent. | my_set.remove(item) |
| .update() | Adds multiple elements in-place from an iterable (e.g., set, list, tuple). | my_set.update(sequence) |

* EXcept, sorted, other methods above modify the set and return `None`, 

**Set Operations**

| Operation | Symbol | Method | Description | Example |
| :--- | :--- | :--- | :--- | :--- |
| Union | `\|` | `.union()` | Combines elements from both sets, returning a new set containing all unique values | `set1 \| set2` |
| Intersection | `&` | `.intersection()` | Returns a new set containing only the elements present in both sets simultaneously | `set1 & set2` |
| Difference | `-` | `.difference()` | Returns a new set containing elements present in the first set but absent in the second | `set1 - set2` |
| Symmetric Difference | `^` | `.symmetric_difference()` | Returns a new set containing elements in either set, excluding any elements present in both | `set1 ^ set2` |
| Subset | `<=`| `.issubset()`| Returns a `True` when set on L.H.S  is a subset of the other on R.H.S | `set1 <= set2` |
| Superset | `>=` | `.issuperset()` | Returns `True` when the set on the L.H.S. is a superset of the one on the R.H.S. | `set1 >= set2` |
| Disjoint | N/A | `.isdisjoint()` | Returns `True` if two sets have a null intersection (no elements in common) | `set1.isdisjoint(set2)` |

```python
# Tuple creation 
tuple1 = (1,) # One element tuple
Not_tuple = (1) # Not a tuple, just an integer
tuple2 = 19,7,1,9,2,91,19 # Multiple element tuple
print("tuple1 is", type(tuple1),"Not_tuple is", type(Not_tuple), "And tuple2 is", type(tuple2))
```

```python
# Tuple methods
print('original tuple is ', tuple2, 'and is of datatype', type(tuple2)) #printing the tuple
print('tuple sorted to give :', sorted(tuple2),'but this is of datatype', type(sorted(tuple2)))
print('index of the first 19 is', tuple2.index(19), '19 appears', tuple2.count(19), 'times')
```

```python
# Tuple unpacking
Arsenal, Barca, Bayern =  ('EPL', 'La Liga', 'Bundesliga') # Variables on LHS must be equal to values in the tuple on the L.H.S
print ('Arsenal plays in: ', Arsenal, 'but Barca plays in', Barca, 'while Bayern plays in', Bayern ) #  In line 2 above, unpacking assigns Arsenal to EPL and Barca to La Liga
```

```python
# Sets
seta={1,1,1,2,3,7,4,4,5,5,6,6,6,6,6,7,7,8,9}
setb={1,2,3,4,5,6,7,8,9}
print('Since sets automatically delete duplicates, seta is equal to setb', seta==setb)
```

```python
# Set methods
setz = set([1,2,5,3,4])
print( 'setz', setz)
print('setz is sorted: ', sorted(setz),'but is of type: ', type(sorted(setz)))
setz.add(109) #Add 9 to setz, Returns None, only works for one object
print('setz is ', setz)
setz.update([6,7,8])# Adds a sequence to setz
print('setz is: ', setz)
setz.discard(109) # removes 9 from setz
print('setz is: ', setz)
```

```python
# Set operations
a ={1,2,3,4,5,6,7,9}
b={0,2,4,6,8}
print('a intersection b is: ', a.intersection(b))
print('a-b(a only) is: ', a.difference(b))
print('b-a(b only) is: ', b-a)
print('a union b is: ', a.union(b))
print("a is subset of", a<=b)
```
