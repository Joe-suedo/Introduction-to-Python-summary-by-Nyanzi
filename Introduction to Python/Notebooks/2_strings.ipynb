## STRINGS; BASICS AND FORMATTING
* Strings: text or character data
* They are  sequential collections of zero or more letters, numbers and other symbols.
* Must be quoted with single `''` or double `""` quotes but the types of quotes must match. 
* class denoted as `<class 'str'>`

### String Operations
* Concatenation; Addition of strings with the `+` operator, leaves no space between strings.
* Multiplication; Repeats the string.
* `in` operator checks for presence of a character in a string.
* `len` returns the number of characters in a string:
  
### Properties of strings
* Ordered: Can  be indexed or sliced, note that python indexing starts at 0
* immutable:character at a particular index cannot be modified
  
**String comparision; follows lexicographical order**
* All uppercase letters are smaller than lowercase letters. For example, "Z" is smaller than "z"
* If number strings,a dictionary pattern is followed instead of a numeric one e.g 10 < 9
* all strings starting with numbers are smaller than strings starting with letters. 

  
### String Formatting
* Creation of strings with placeholders that insert the values assigned to the provided variables in the output.
* Placeholders are initiated with the % operator followed by format specifiers that dictate data type and presentation.
  
| Placeholder | format |
| ----------- | ------ |
| `%s` | string |
| `%d` | interger |
| `%f` | float |
| `%o` | octal |
| `%x` | hexadecimal |
| `%e` | exponential |

### String methods
* Invoked by appending a dot character (`.`) immediately after a string variable or a direct string literal.
* This is followed by the specific method name and parentheses containing any required arguments.

| Method | Description |
| :--- | :--- |
| .count(x) | Returns the number of occurrences of substring `x` in the string. |
| .capitalize() | Converts the first character of a string to uppercase. |
| .upper() | Converts all characters in the string to uppercase. |
| .lower() | Converts all characters in the string to lowercase. |
| .replace('a', 'b') | Replaces occurrences of substring `'a'` with `'b'`. |
| .strip() | Removes leading and trailing whitespace from the string. |
| .rstrip() | Removes trailing (right-side) whitespace. |
| .lstrip() | Removes leading (left-side) whitespace. |
| .ljust(width) | left-justified in a field of specified size width |
| .rjust(width) | right-justified in a field of specified size width |
| .split(sep) | Splits the string into a list of substrings using a specified separator (defaults to whitespace). |
| .splitlines() | Splits the string at line breaks into a list of lines. |
| .center(width) | Centers the text within a string of a specified  width. |
| .find() | Searches for a substring and returns the first index where it begins, or `-1` if not found. |
| .join(list) | Joins the elements of a given list together using the string as a delimiter. |
| .isalpha() / .isdigit() / .isspace() | Tests if all string characters belong to specific character classes. |
| .startswith() / .endswith() | Tests if the string starts or ends with a given substring. |

```python
print('Nya'+'nzi') # Concatenation; note that no space
print('My name is' + ' ' + 'Nya'+'nzi') #Note that a space is added also enclosed in quotes as a string.
'paw'*2 #Multiplication; note that the string is repeated twice without any space in between.
```

```python
# Lexicographical order of strings; Comparision of strings.
print('1a is smaller than a1', '1a' < 'a1')
print('1a is smaller than 1b', '1a' < '1b') 
print('A is smaller than a', 'A' < 'a')
print('1000 is smaller than 9', '1000' < '9')
print('bird is smaller than z', 'bird' < 'z')
```

```python
#String formatting
cost = 100
new_cost = 150
print("The cost of sweets was", cost, "shs", "but now it is", new_cost, "shs") # No string formatting
print("The cost of sweets was %d shs but now it is %d shs" % (cost, new_cost)) # Using string formatting 
```

```python
# String formatting
faith = 'I am a catholic and i love Mother Mary'
print ('catholic is in faith', 'catholic' in faith) #note that it returns a boolean value.
print ('I went to %s for undergaduate and %s for masters' % ('MUST', 'Mak')) #String formatting.
print('I will have 1%ebillion dollars' %50)
```

```python
#Indexing Strings
name = 'Nyanzi'
print(name[0]) # Accessing the first character of the string
print(name[0:3]) # Accessing the first three characters of the string
#name[0] = 'n' # Strings are immutable, this will raise an error
sub_name = name[0:3] # Concatenating strings to create a new string, this is allowed
print(sub_name) # Printing the new string
print('y is present in name, name is Nyanzi', 'y' in name)  # Checking if  character n is in str name
```
