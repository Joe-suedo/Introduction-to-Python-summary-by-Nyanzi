## Loops; `while` and `for`
* Necessary when we have to perform a set of functions repeatedly.
* Code to be executed when condition is satisfied must be indented. 
* Indentation accomplished with the 'tab' key which is equivalent to typing two spaces or four spaces.
* Expression following the 'while' must end with a colon (:)
  
| While loops | For loops |
| :---------- | :-------- |
| executes a code block when the loop's expression is `True` | executes a block of code n times |


**while loop**
* checks a logical condition and execute code block until the condition being checked becomes false.
* Infinite `while` loop : always execute because the loop's expression is always `True`.

**for loop**
*  useful when we know how many times a loop is to be executed.
*  controls the number of times a loop will iterate by counting or stepping through a series of values.
*  `range`: command creates a series of five values from 0 to 4
*  

```python
# while loop
score = int(input("Enter student's score below or equal to 100:"))
while (score > 100):
    print('%i is over 100, student cannot score above 100, ' % score)
    score = int((input('Please enter a number less than 100: ')))
print('Thank you, %i is the final score of the student' % score )
```
