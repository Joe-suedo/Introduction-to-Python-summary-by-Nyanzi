# FILE HANDLING

**File input and output (or File I/O):** ability to read data from and write data to files stored in directories/folders.

**Major steps to referencing a file**
| Step | Syntax |
| :--- | :--- |
| Open the file <br>Tells the operating system the name of the file & it's location and how to use it| `open(filename or path, mode):` <br>'r' to read data in the file (assumed  by default if no option added)<br> 'w' to write to file <br> 'a' to add/append data to file|
| Perform operations on the file data. ||
| Close the file. |filename.close() |

* The `open()` function initializes a text wrapper object that enables navigation, reading, and file manipulation[cite: 7].

**Methods of Reading from Files**

| Method / Approach | Description | Example |
| :--- | :--- | :--- |
| `read()` | Reads everything from a file and returns it as a single string | `read_file.read()` |
| `readline()` | Reads one line from a file and returns it as a string, advancing the cursor line by line. | `read_file.readline()` |
| `readlines()` | Returns a list of strings containing all the lines from the file. | `read_file.readlines()` |
| `for i in f` | Iterates over the lines in a file sequentially, where `i` represents an individual line. | `for i in read_file:` |

**Methods for Writing Files**

| Method / Mode | Description | Example |
| :--- | :--- | :--- |
| `.write()` | Writes a single string to the file (requires explicit newline characters like `\n` for separation). | `file_obj.write("text\n")` |
| `.writelines()` | Writes an iterable or list of strings to the file back-to-back without automatic separators (requires explicit newline characters like `\n` for line breaks). | `file_obj.writelines(["line1\n", "line2\n"])` |

**File Navigation Methods**

| Method | Description | Example |
| :--- | :--- | :--- |
| `seek()` | Moves the file cursor to a specified byte position; for example, `seek(0)` brings the cursor back to the start of the file[cite: 7]. | `read_file.seek(0)`|

### File Management: `open()` vs. Context Manager (`with open()`)

| Approach | Syntax & Usage | Description & Best Practice |
| :--- | :--- | :--- |
| **Traditional (`open()`)** | `f = open('file.txt', 'r')`<br>`content = f.read()`<br>`f.close()` | Requires explicit `.close()` calls. If an exception occurs before reaching `.close()`, the file resource remains open, leading to potential resource leaks. |
| **Context Manager (`with open()`)** | `with open('file.txt', 'r') as f:`<br>`    content = f.read()` | **Recommended Best Practice.** Automatically closes the file as soon as the indented block exits, guaranteeing proper cleanup even if errors occur. |
