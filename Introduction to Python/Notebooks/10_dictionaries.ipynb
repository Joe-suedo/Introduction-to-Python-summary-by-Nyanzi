# Dictionaries

Dictionaries represent mapping structures that store associations between unique, immutable keys and their corresponding values, forming key-value pairs (`key:value`).

### Properties of Dictionaries

| Property | Description |
| :--- | :--- |
| Mapping Structure | Associates unique keys to values using key-value pairs. |
| Mutable Container | The dictionary container can be modified dynamically; entries can be added, updated, or removed after creation. |
| Mutable Values | Values associated with keys can be referenced, replaced, or modified freely, and can be of any data type (including lists or other dictionaries). |
| Immutable Keys | Keys must be immutable and hashable (e.g., strings, numbers, tuples); attempting to use a mutable object like a list as a key raises a TypeError. |
| Insertion Ordered | Starting from Python 3.7, dictionaries maintain the insertion order of their keys. |

### Creation of Dictionaries

| Approach / Method | Syntax | Description | Example |
| :--- | :--- | :--- | :--- |
| Literal or Constructor | `{}` or `dict()` | Initializes an empty dictionary. | `my_dict = {}` |
| Key-Value Assignment | `[key] = value` | Adds a new key-value pair or updates an existing one. | `my_dict['key'] = 'value'` |
| Zip Function | `dict(zip())` | Constructs a dictionary by pairing parallel iterables of keys and values. | `dict(zip('abc', range(3)))` |
| Dictionary Comprehension | `{k: v for ...}` | Generates a dictionary dynamically using an iterative expression. | `{x: x**2 for x in range(3)}` |

### Dictionary Lookups

| Lookup Type | Description | Example |
| :--- | :--- | :--- |
| Direct Lookup | Retrieves the value associated with a key; raises a KeyError if the key is missing. | `my_dict[key]` |
| Safe Lookup | Retrieves the value for a key safely, returning None (or a specified default) instead of raising an error if absent. | `my_dict.get(key, default)` |
| Reverse Lookup | Custom iteration logic required to find a key given a value (performs a slower linear scan). | `def reverse_lookup(d, v): ...` |

### Operations and Functions

| Operation / Function | Description | Example |
| :--- | :--- | :--- |
| `in` operator | Checks if a key exists in the dictionary, returning a boolean (`True` or `False`) in constant time. | `key in my_dict` |
| `sorted()` function | Returns a new sorted list containing the dictionary's keys (requires comparable key types). | `sorted(my_dict)` |
| `del` statement | Deletes a specified key and its associated value from the dictionary. | `del my_dict[key]` |
| For Loop | Iterates through the dictionary's keys sequentially. | `for key in my_dict:` |
| Merge Operator (`\|`) | Python 3.9+ operator that merges two dictionaries, returning a new dictionary. | `merged_dict = dict1 \| dict2` |

### Dictionary Methods

| Method | Description | Example |
| :--- | :--- | :--- |
| `.keys()` | Returns a dynamic view object of all keys in the dictionary. | `my_dict.keys()` |
| `.values()` | Returns a dynamic view object of all values (combining this view with the `in` operator enables value existence checks). | `value in my_dict.values()` |
| `.items()` | Returns a dynamic view object of key-value pairs formatted as tuples. | `my_dict.items()` |
| `.update()` | Merges key-value pairs from another dictionary or iterable of key-value pairs in-place (or via `\|=`). | `my_dict.update(other_dict)` |
| `.pop()` | Removes a specified key and returns its value, raising a KeyError if the key is missing unless a default is provided. | `my_dict.pop(key, default)` |
| `.setdefault()` | Returns the value of a key if it exists; if not, inserts the key with a specified default value. | `my_dict.setdefault(key, default)` |
| `.clear()` | Removes all items from the dictionary, leaving it empty. | `my_dict.clear()` |

### Technical Architecture Notes

- Built on optimized hash table data structures, allowing search, insertion, deletion, and key lookup operations to execute in average constant time ($O(1)$) regardless of dictionary size.
- Keys rely on hash values computed by Python's hash function; therefore, any object used as a key must be hashable and immutable.
- Lists and other dictionaries cannot serve as keys because their mutability prevents consistent hashing, but they are fully permitted as values.

```python
#Creation of a dictionary using the zip() function
marks = ["100-90", "89-70", "69-50", "49-30", "29-10", "9-0"]
grades = ["A", "B", "C", "D", "E", "F"]
grade_dict = dict(zip(marks, grades))
print('grade_dict:', grade_dict)
print('grade_dict has a data type of:', type(grade_dict))

#Dictionary comprehension 
squares={x: x*x for x in range(1,11)}
print('squares:', squares)
print('squares has a data type of:', type(squares))
```

```python
# Merging two dictionaries using the update() method
dict1 = {'a': 1, 'b': 2}
keys_2 = ['c', 'd']
values_2 = [3, 4]
dict1.update(dict(zip(keys_2, values_2)))
print('dict1 after merging:', dict1)
```

```python
# Mutability of dictionaries
print('Before modification:', dict1)
dict1['a'] = 10 #Modifying the value of an existing key, key can not be modified
del dict1['b'] #Deleting a key-value pair from the dictionary
dict1['e'] = 5 #Adding a new key-value pair to the dictionary
print('After modification:', dict1)
```
