## if, else, and elif Statements
* Control the logical flow of the program.

**Syntax**
* "if" must end with a colon (:) 
* The code to be executed when the logical condition is true must be indented.
* Indentation accomplished  with the "tab" key which is equivalent to typing two or four spaces depending on editor.
  
**`if` statement**
* Executes an indented block of code only when the condition evaluates to `True`.
* Skipped entirely when the logical condition is not satisfied.
  
**`if-else` statement**
* Allows for two separate code blocks
* one executed when the condition is true
* and an alternative block executed when it is false.
  
**`if-elif` Statement**
* Evaluates multiple conditions sequentially from top to bottom.
* Executes **only** the code block belonging to the first condition that evaluates to `True`.
  
**Multiple `if` statements**
* Evaluates every condition independently from top to bottom.
* Executes **every** code block whose condition evaluates to `True`, unlike `if-elif` which stops after the first match.
  
**Nested `if` statement**
* Places an `if` statement inside another `if` block.
* Evaluates the inner condition only when the outer condition is satisfied (`True`).

```python
#if statement
avalue= int(input('Enter an integer greater than 10: '))
if avalue>10:
    print("The avalue %i is greater than 10 " %  avalue)
```

```python
# if-else statement
bvalue= int(input('Enter an integer greater than 10: '))
if bvalue>10:
    print("The bvalue %i is greater than 10 " %  bvalue)
else:
    print("The bvalue %i is not greater than 10 " %  bvalue)
```

```python
# if-elif-else statement
cvalue= int(input('Enter an integer greater than 10: '))
if cvalue>10:
    print("The cvalue %i is greater than 10 " %  cvalue)
elif cvalue==10:
    print("The cvalue %i is equal to 10 " %  cvalue)
else:
    print("The cvalue %i is less than 10 " %  cvalue)
```

```python
#Multiple if statements-More than one code block executed if more than one condition is true.
test_score = float(input("Enter test score:"))
if test_score >= 90:
 print("%.2f is an A grade!" % test_score)
if 80 <= test_score < 90:
 print("%.2f is a B grade!" % test_score)
if 70 <= test_score < 80:
 print("%.2f is a C grade!" % test_score)
if 60 <= test_score < 70:
 print("%.2f is a D grade!" % test_score)
if test_score < 60:
 print("%f is an F grade!" % test_score)
```

```python
# Nested `if statement`
grade = float(input("Enter your grade: "))
if grade >= 90:
  if grade < 95:
    print("%.2f is an A-" % grade)
  elif grade >= 95:
    print("%.2f is an A+" % grade)
else:
    print("%.2f is not an A grade" % grade)
```
