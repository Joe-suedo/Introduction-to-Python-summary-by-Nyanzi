## FUNCTIONS
* Named sequence of statements that performs a computation
* For execution of a code over and over again as it  allows the reuse of code by calling it.
  
**Uses of functions**
* User defined functions are easier to read and debug.
* Make a program smaller by eliminating repetitive code. 
* Can be modelled to solve a specific problem.
* Well-designed functions are often useful for many programs.

| Category | Definition & characteristics | Examples |
| :--- | :--- | :--- |
| Built-in functions | pre-installed with Python installation & globally available for use without importing any modules.| print(), len(), type(), range() |
| Methods | Functions associated with specific objects and  called directly on that object using dot notation. | append(), .upper(), .split() |
| User-defined functions | Custom functions created and defined by programmers using the def keyword to perform reusable tasks | - |
| Module |• A file containing a collection of related functions <br>• Loaded using the `import` keyword followed by the module name <br>• Functions inside a module are accessed using dot notation <br>(module name followed by a dot and the function name) | • math, numpy, pandas <br>• import math <br>• math.sqrt(16) |

### User-defined functions
* function definition: specifies the name of a new function and the sequence of statements/expressions that run when the function is called.

**Syntax for User-defined functions**
| Part of function definition | Description |
| :--- | :--- |
| Header |. First line of function definition<br>. Starts with `def` to declare a function definition <br>. Followed by the function name assigned by the user<br>. Followed by parameters in parantheses[empty parantheses`()` if no arguments required]<br>. Ends with a colon`:` |
| Body | • Indented block of code (typically 4 spaces) following the header<br>• Contains the statements and logic executed when called<br>• Optionally ends with a `return` statement to output values back to the caller |

**Parameters vs Arguments**
* Arguments(values) are assigned to parameters(variables)
  
| Parameter | Argument |
| :--- | :--- |
| Variables used in function definition | values used in function calls |

**Local vs Global Variables**
| Local Variable | Global variable |
| :--- | :--- |
| Declared inside a function and accessible only within that function's scope | Declared outside any function or class and accessible throughout the program |

**Stack Diagram**
* Visualizes each active function as a separate frame (box) labeled with the function's name.
* Displays all parameters and local variables belonging to that function along with their current values inside the box.
  
**Traceback**
* A report detailing the active functions when an error occurs.
* Identifies the program file, specific line number, and exact line of code that triggered the error.
* Traces the sequence of active function calls leading up to the failure.

**Fruitful vs Void Functions**
| Fruitful functions | Void functions |
| :--- | :--- |
|`return` a value that can be stored in a variable, printed, or passed into another calculation.| Perfom an action like printing text to the console, modifying a list in-place, or writing to a file) but implicitly `return None`|

**Notes**
* An empty line ends a function in iterative mode.
* A defined function can be used inside another function. 
* Function definition has to run before the function gets called.
* Statements are run one at a time, in order from top to bottom.


```python
# User defined function to greet a user
def greet_user(name):
    """Display a simple greeting."""
    print(f"Hello, {name}!")
```

```python
# Call the function with a name
greet_user("Alice")
```
