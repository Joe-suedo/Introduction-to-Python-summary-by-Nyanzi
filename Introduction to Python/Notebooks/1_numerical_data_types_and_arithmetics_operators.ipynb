## NUMERIC DATA TYPES AND ARITHMETIC OPERATORS : 
**Basic syntax for statements**
* No spaces or tab characters allowed at the start of a statement.
* The '#' character indicates that the rest of the line is a comment
* Statements finish at the end of the line:
* Except when there is an open bracket or parenthesis(:) or back slash
  
**Numeric Data types**
| int | float |
| --- | ----- |
| <class 'int'> | <class 'float'> |
| Integer(whole) number | floating-point(decimal) number |
| No decimal points;whole number  | Has decimal points or scientific notation |
| For example: 3, 4, 894, 0, -3, -18 | For example: 5.6, 7.342, 53452.0, 0.0, -89.34, -9.5 |
 
 * if at least one of the operands in an arithmetic operation is a float, the result will typically be a float. 

#### Arithmetic operators
| Operator | Name | Description |
| -------- | ---- | ----------- |
| + | Addition ||
| - | Substraction ||
| * | Multiplication ||
| ** | Exponenation | Raises the left number to the power of the right |
| / | Division | returns float even if the numbers divide without a remainder |
| // | Floor division | truncates the decimal and returns an int |
| % | modulo | Returns the remainder,Usefulfor checking even/odd numbers. Finds numbers that are multiples of the same number.|

* Operand : value on which an operator operates.
* Operators : functionality that do something, represented by symbols or by special keywords.

**Arithmetic Operator Precedence**
* Multiple operators in a single expression are evaluated in order of precedence.
* Associativity:Operators of the same precedence are evaluated from left to right.
   
| Operator | Precedence |
| -------- | ---------- |
| ** | highest |
| - (negation) ||
| *, /, //, % ||
| + (addition), - (subtraction) | lowest |

* Alternatively, use PEMDAS
  
| Shorthand | Operator |
| --------- | -------- |  
| P | Paranthesis |
| E | Exponation |
| M & D | Multiplication & Division |
| A & S | Addition & Subtraction |
| For operators of the same precedence | evaluated in a left to right manner|

**Unary Arithmetic Operations**
 * Unary mathematical expression: consists of only one component or element
 * Plus`+` & minus`-` signs are used as a single element paired with a value to return the value's identity ;Negative or Positive
  
#### Variables and Assignment Statements

*Variable:* a name that refers to a value, values assigned to variables can change throughout computation.
*Value:* data to be stored in a computer.
* The assignment operator `=` links a variable name to a given value.
  
**Variable name syntax**
 * Variable names are case-sensitive.
 * Reserved words carry built-in meanings and cannot be used as variable names.
 * Reassigning an existing variable name to a new value replaces and discards its previous value.
 * Long, descriptive variable names are preferred over short ones.
 * Can contain both letters and digits, but they have to begin with a letter or an underscore.
 * Multi-word variable names can use underscores (`_`) instead of spaces
 * Uppercase letters are typically not used.
 * Alternatively, *camelCaps*(starting with a lowercase letter and capitalizing each subsequent word).
 * Code is read and maintained by humans, not solely executed by machines; add comments after hash `#` character
  
**Assignment Operators**
 * Equal sign `=` : assigns the value on the right to a variable on the left. 
 * For example, v = 23: assigns the value of the integer 23 to the variable v.
*Compound Assignment Operators*
  * Modify a variable's existing value using an operation and store the updated result back into that same variable.
  * combine an arithmetic operator with the `=` operator and commonly used with loops
  
  | Compound Assignment Operator | Description |
  | ---------------------------- | ----------- |
  | `+=` | Addition |
  | `*=` | Multiplication |
  | In short | var = var operation expression becomes var operation= expression |

```python
# Demonstration of Numerical Data Types in Python and Print command
a = 7 # the value 7 is assigned to the variable named a
b = 7.0 # the value 7.0 is assigned to the variable named b
c = 7,000 #  this will create a tuple instead of a number,commas are not used to seperate thousands.
print("Type of a:", type(a))
print("Type of b:", type(b))
print('c is', c) # A tuple, not a number 
```

```python
# Demonstration of operators with Nauces
print("Division,returns a float:7/7 gives", 7 / 7)
print("Floor Division,truncates the decimal part, gives integer:7//2 gives", 7// 2)
print("Modulus,returns the the whole number integer remainder:7%2 gives", 7 % 2)
```

```python
# Division operators with negative values
print(-10/3)  # Same as dividing non negative intergers.
print(-10//3) #Floor division returns the largest integer less than or equal to the result of the division. 10/3 gives -3.3333, the largest integer less than or equal to -3.3333 is -4 
print(-10%3) #Evaluated as (remainder a/b) = a - (a  //  b)* b, result takes the sign of the divisor.
```

```python
# Demonstration of unary operations
x = 5
y = -7
print("Unary operation, negation of x: -x gives", -x)
print("Unary operation, negation of y: -y gives", -y)
```

```python
#Demonstration of reassigning variables
x = 10
x = 20
print("Reassigned value of x:", x)
```

```python
# Demonstration of Compound Assignment Operators
x = 5
x += 3  # equivalent to x = x + 3
print("After compound assignment x += 3, x is now:", x)
```

```python
# Demonstration of statements and expressions
y = 3 # This is a statement 
x = y**2  # This is a statement
print("The value of y in memory is:", y, "and the value of x in memory is:", x)
print("The value of x**2 is:", x**2, "but it is not in memory")  # This is  an expression
```
