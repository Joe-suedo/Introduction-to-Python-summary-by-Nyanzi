## Bool Datatype
* Necessary for the evaluation of logical data.
* `Bool` stands for Boolean data, which can only take on one of two possible values: `True`or `False`.
* `Boolean`,`True` and `False` always begins with a capital letters. as they are special characters.
  
**Uses of `Bool`**
* For making comparisons
* Controlling the flow of the program.

**Other uses of `bool`**

*Arithmetic Integration:*
* Booleans act as numeric subtypes where `True` evaluates to `1` and `False` evaluates to `0`.
  
*Membership Operator (`in`):* 
* Checks for the presence of an element or character within a sequence (such as strings or lists) and returns a boolean.
  
*Identity Operator (`is`):* 
* Checks whether two variables point to the exact same object in memory and returns a boolean.
  
### Relational Operators; For making comparisions
* Compare values in order to determine their relationship
* Possible answers are only `True` or `False`
  
| Operator | Description |
| -------- | ----------- |
| == | Equals |
| != | Not Equals |
| > | Greater than |
| >= | Greater than or Equal To |
| < | Less than |
| <= | Less than or Equal To |

* Caution!: `=` is an assignment operator but `==` is a relational operator
* Python compares the values not the data type when working with relational operators.
  
| Example | Description |
| ------- | ----------- |
| x `=` 5.0 | Assigns the value 5.o to variable called x |
| x `==` 5.0 | Evaluates wether the value assigned to variable name x is equal to 5.0 |

**Logical Operators**
* Relational operators can be chained with logical operators.
  
| Operator | Alt(Symbol) | `True` when |
| -------- | ----------- | ----------- |
| x `and` y | `&` | Both x and y are `True` |
| x `or`  y | \| | Atleast one condition is `True` |
| `not` x | N/A | x is `False` |

* Compound statements can be written using `and`, `or`, and `not`

*Comparision of Strings*
* They are case-sensitive unless you employ an additional string method.

**Truth tables**
* Make decision making faster
* Example of `==` truth table, other truthful tables not shown.
  
| x | `==` | y | Results |
| - | ---- | - | ------- |
| `True` | `==` | `True` | `True` |
| `True` | `==` | `False` | `False` |
| `False`| `==` | `True` | `False` |
| `False` | `==` | `False` | `True` |

**Operator Chaining**
* Operators in python can be chained together
* Chaining performs comparisons left to right
    *For instance a>b>c*
    * Python determines b is greater-than a but less-than c.
    * `a>b` is evaluated first.
    * when `a>b` is `True` then `b>c` is evaluated.
    * `b>c` is not evaluated at all if `a>b` is `False`

### Boolean operators for flow control.
* Boolean: control the stream and outcomes of a program in flow statements.
* Condition: Evaluates to `True` or `False`, presenting a point for decision making in the program.
* Clause: block of code that follows the condition and dictates the outcome of the program





```python
# Demonstration of other uses of Bool
print('True + 1 + False',True+ 1 + False) # gives 2 as True is 1 and False is 0
x=(1,2,3,4)
y=x
print('1, a member of x', 1 in x) # checks membership
print('y same as x', y is x)
```

```python
# Demonstration of use of bool data type to make comparisions
a = 3
b = 2
c = 1
print("a is greater than b", a>2)
print("b is less than c", b<c)
print("a is equal to b plus c", a==b+c)
print("a is not equal to b", a!=b)
print("b is greater  than or equal to 2", b>=2)
# Chaining the comparisons(not, and, or)
print("a is greater than b and b is greater than c", a>b and b>c)
print("a is greater than b or c is greater than b", a>b or c>b) # Only one condition is True
print("a is not equal to b and b is not equal to c", not(a==b) and not(b==c)) # Both conditions are True
```
